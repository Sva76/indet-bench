```python
# 7b. Ambienti uv separati, repository degli autori e sonde alle revisioni fissate (come nel notebook A del Test M;
# i dataset annotati del controllo positivo di M4 qui non servono, perché HC0 si eredita dal Test M).
import shutil

UV_VERSION = "0.11.32"
ENVS = LAVORO / "envs"
SCRIPT = LAVORO / "script"                 # gli script tm_*.py e tc_*.py sono scritti dalle celle 8
LOG = RADICE / "log"
AMBIENTI = RADICE / "ambienti"
for d in (ENVS, LOG, AMBIENTI):
    d.mkdir(parents=True, exist_ok=True)
SIRIN_DIR, OBESO_DIR, HF_DIR = LAVORO / "SIRIN", LAVORO / "hallucination_probes", LAVORO / "hf"
PY = {"env_m1": ENVS / "m1/bin/python", "env_sirin": ENVS / "sirin/bin/python", "env_obeso": ENVS / "obeso/bin/python"}

def sh(cmd, log, cwd=None, env=None):
    # esegue un comando, mostra le ultime righe e salva tutto nel log su Drive
    with open(log, "a") as fh:
        fh.write(f"\n$ {cmd}\n")
        fh.flush()
        p = subprocess.run(cmd, shell=True, cwd=cwd, env=env, stdout=fh, stderr=subprocess.STDOUT)
    if p.returncode != 0:
        tail = Path(log).read_text()[-4000:]
        raise RuntimeError(f"comando fallito ({p.returncode}): {cmd}\n--- ultime righe del log ---\n{tail}")

def git_checkout(url, folder, commit, log):
    if not (folder / ".git").exists():
        sh(f"git clone -q {url} {folder}", log)
    sh(f"git -C {folder} fetch -q origin {commit} || true", log)
    sh(f"git -C {folder} checkout -q {commit}", log)
    head = subprocess.run(["git", "-C", str(folder), "rev-parse", "HEAD"], capture_output=True, text=True).stdout.strip()
    assert head == commit, f"{folder.name}: commit {head} invece di {commit}"

PACCHETTI_EFFETTIVI = {}                    # impronta dei pacchetti letti dall'ambiente installato, in questa sessione

def ensure_uv():
    if shutil.which("uv") is None:
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", f"uv=={UV_VERSION}"], check=True)

def freeze_text(name):
    # elenco dei pacchetti dell'ambiente così com'è adesso, letto dal suo Python
    ensure_uv()
    p = subprocess.run(["uv", "pip", "freeze", "--python", str(PY[name])], capture_output=True, text=True)
    if p.returncode != 0 or not p.stdout.strip():
        raise RuntimeError(f"{name}: uv pip freeze non riuscito\n{p.stderr[-2000:]}")
    return p.stdout

def freeze(name, log):
    out = freeze_text(name)
    (AMBIENTI / f"{name}.txt").write_text(out)
    PACCHETTI_EFFETTIVI[name] = fingerprint(out)
    return PACCHETTI_EFFETTIVI[name]

def pin_fp(name):
    # impronta di ciò che definisce l'installazione: le versioni fissate (o il commit col lock degli autori)
    return fingerprint({"env_m1": PIN_M1, "env_sirin": PIN_SIRIN + SIRIN_COMMIT, "env_obeso": OBESO_COMMIT}[name])

def env_ready(name):
    return (ENVS / f"{name}.ok").exists() and PY[name].exists()

def setup_env(name):
    log = LOG / f"installazione_{name}.log"
    if env_ready(name):
        # Ambiente già installato: si confrontano i pacchetti effettivi con l'installazione e con l'elenco su Drive.
        info = read_json(ENVS / f"{name}.ok")
        actual = freeze_text(name)
        saved = AMBIENTI / f"{name}.txt"
        if info.get("pin") != pin_fp(name):
            # le versioni fissate sono cambiate (nuova versione del notebook): si reinstalla da zero
            print(f"{name}: versioni fissate cambiate rispetto all'installazione. Reinstallo l'ambiente.")
            shutil.rmtree(PY[name].parent.parent)
            (ENVS / f"{name}.ok").unlink()
        elif fingerprint(actual) != info["pacchetti"]:
            # l'ambiente è cambiato dopo l'installazione: si reinstalla da zero con le versioni fissate
            with open(log, "a") as fh:
                fh.write(f"\n{name}: pacchetti diversi dall'installazione, reinstallo\n")
            print(f"{name}: i pacchetti installati non coincidono con quelli dell'installazione. Reinstallo l'ambiente.")
            shutil.rmtree(PY[name].parent.parent)
            (ENVS / f"{name}.ok").unlink()
        else:
            if saved.is_file() and saved.read_text() != actual:
                # l'elenco su Drive viene da un altro ambiente: si conserva, ma vale quello effettivo, e i lavori
                # fatti con l'altro ambiente non hanno più l'impronta attesa
                saved.rename(saved.with_name(f"{name}.precedente.{fingerprint(saved.read_text())[:12]}.txt"))
                print(f"{name}: l'elenco dei pacchetti su Drive era di un altro ambiente; i lavori fatti con quello "
                      "non si riusano.")
            saved.write_text(actual)
            PACCHETTI_EFFETTIVI[name] = fingerprint(actual)
            if name == "env_obeso":              # sonde e dataset: si riscaricano (già presenti = istantaneo) e si verificano
                download_obeso_assets(log)
            return info
    sh(f"{sys.executable} -m pip install -q uv=={UV_VERSION}", log)
    folder = PY[name].parent.parent
    if name == "env_m1":
        (LAVORO / "pin_m1.txt").write_text(PIN_M1)
        sh(f"uv venv -q --python 3.12 {folder}", log)
        sh(f"uv pip install -q --python {PY[name]} -r {LAVORO / 'pin_m1.txt'}", log)
    elif name == "env_sirin":
        git_checkout("https://github.com/sb-ai-lab/SIRIN.git", SIRIN_DIR, SIRIN_COMMIT, log)
        (LAVORO / "pin_sirin.txt").write_text(PIN_SIRIN)
        sh(f"uv venv -q --python 3.12 {folder}", log)
        sh(f"uv pip install -q --python {PY[name]} -r {LAVORO / 'pin_sirin.txt'}", log)
    elif name == "env_obeso":
        git_checkout("https://github.com/obalcells/hallucination_probes.git", OBESO_DIR, OBESO_COMMIT, log)
        env = dict(os.environ, UV_PROJECT_ENVIRONMENT=str(folder))
        sh("uv sync -q --frozen --python 3.11", log, cwd=OBESO_DIR, env=env)
        download_obeso_assets(log)
    info = dict(ambiente=name, pacchetti=freeze(name, log), pin=pin_fp(name))
    write_json(ENVS / f"{name}.ok", info)
    print(f"{name}: pronto (impronta dei pacchetti {info['pacchetti'][:12]}, elenco in {AMBIENTI / (name + '.txt')})")
    return info

def download_obeso_assets(log):
    from huggingface_hub import snapshot_download
    token = hf_token()
    for probe_id in SONDE:
        snapshot_download(SONDE_REPO, revision=SONDE_SHA, allow_patterns=[f"{probe_id}/*"],
                          local_dir=OBESO_DIR / "value_head_probes", token=token)
        assert (OBESO_DIR / "value_head_probes" / probe_id / "probe_head.bin").exists(), probe_id

_HF_TOKEN = []                              # letto una volta sola dai Secrets di Colab, solo in memoria

def hf_token():
    # Colab a volte non risponde alla richiesta del Secret durante una cella lunga: si legge all'inizio, con tre
    # tentativi, e poi si riusa. Il token non si scrive mai su file né si stampa.
    if _HF_TOKEN:
        return _HF_TOKEN[0]
    token = os.environ.get("HF_TOKEN")
    for attempt in range(3):
        if token:
            break
        try:
            from google.colab import userdata
            token = userdata.get("HF_TOKEN")
        except Exception:                        # noqa: BLE001
            token = None
            time.sleep(2)
    if token:
        _HF_TOKEN.append(token)
    return token

print("Token Hugging Face:", "presente" if hf_token() else "ASSENTE (Llama e i modelli protetti non si scaricano)")
```

```python
%%writefile /content/testC_run/script/tm_comune.py
# Funzioni comuni agli script del Test C (riprese dal Test M, con il numero di token generati come parametro). Girano in tutti e tre gli ambienti (Python 3.11 e 3.12):
# usano solo la libreria standard, numpy e torch.
import hashlib
import json
import os
import re
import sys
import time
from pathlib import Path

LABELS = ("YES", "NO", "UNKNOWN")


def encode(obj):
    return json.dumps(obj, sort_keys=True, ensure_ascii=False, allow_nan=False).encode("utf-8")


def fingerprint(obj):
    return hashlib.sha256(encode(obj)).hexdigest()


def write_json(path, obj):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    tmp = path.with_suffix(path.suffix + ".tmp")
    tmp.write_bytes(encode(obj))
    os.replace(tmp, path)


def read_json(path):
    return json.loads(Path(path).read_text())


def say(*args):
    print(time.strftime("%H:%M:%S"), *args, flush=True)


def parse_label(text):
    match = re.match(r"""^\s*[*#`"']*\s*(YES|NO|UNKNOWN)\b""", text, re.IGNORECASE)
    return match.group(1).upper() if match else "ALTRO"


def finite(x):
    return x is not None and x == x and x not in (float("inf"), float("-inf"))


def clean(x):
    # valori non finiti -> None, così il JSON resta valido e la coppia si esclude nell'analisi
    x = float(x)
    return x if finite(x) else None


def batches(rows, size):
    for start in range(0, len(rows), size):
        yield rows[start:start + size]


def checkpoint(folder, stage, rows, compute, size=64, key="row_id"):
    # Salva ogni blocco su Drive: una run interrotta riprende dai blocchi già scritti.
    # l'impronta copre il contenuto completo delle righe (prompt compresi), non solo gli identificativi
    sig = fingerprint(rows)[:16]
    folder = Path(folder) / "checkpoint" / f"{stage}-{sig}"
    folder.mkdir(parents=True, exist_ok=True)
    out = []
    for i, chunk in enumerate(batches(rows, size)):
        path = folder / f"{i:04d}.json"
        expected = [r[key] for r in chunk]
        if path.exists():
            part = read_json(path)
        else:
            part = list(compute(chunk, i))
            if [r[key] for r in part] != expected:
                raise RuntimeError(f"Risultati incompleti in {stage}, blocco {i}")
            write_json(path, part)
        if [r[key] for r in part] != expected:
            raise RuntimeError(f"Checkpoint incoerente in {stage}, blocco {i}")
        out.extend(part)
        if i == 0 or (i + 1) % 5 == 0 or len(out) == len(rows):
            say(f"  {stage} {len(out)}/{len(rows)}")
    return out


# ---------- codifica dei prompt e p(UNKNOWN), identiche al Test U2 ----------

def chat_text(tokenizer, model_key, prompt):
    options = {"enable_thinking": False} if model_key == "qwen" else {}
    return tokenizer.apply_chat_template([{"role": "user", "content": prompt}], tokenize=False,
                                         add_generation_prompt=True, **options)


def prompt_ids(tokenizer, model_key, prompt):
    return tokenizer(chat_text(tokenizer, model_key, prompt), add_special_tokens=False)["input_ids"]


def label_ids(tokenizer):
    audit = {}
    for label in LABELS:
        variants = [prefix + word for word in (label, label.capitalize(), label.lower()) for prefix in ("", " ")]
        audit[label] = {v: tokenizer.encode(v, add_special_tokens=False) for v in variants}
    ids = {label: sorted({sequence[0] for sequence in audit[label].values()}) for label in LABELS}
    if len(set().union(*map(set, ids.values()))) != sum(map(len, ids.values())):
        raise RuntimeError(f"Il tokenizer condivide i primi token fra etichette: {ids}")
    return ids


def stop_ids(tokenizer, model):
    ids = set()
    for value in (tokenizer.eos_token_id, getattr(model.generation_config, "eos_token_id", None)):
        if isinstance(value, int):
            ids.add(value)
        elif isinstance(value, (list, tuple)):
            ids.update(v for v in value if isinstance(v, int))
    return ids


def predict_reference(model, tokenizer, model_key, rows, ids, keep_answer_ids=False, max_new_tokens=8):
    # Come la cella 8 del Test U2: batch 1, greedy, probabilità delle etichette sul primo token. Nel Test U2 i token
    # generati erano 8; nel Test C la risposta si genera fino a 32 token (protocollo v3, §4). p(UNKNOWN) dipende solo dal
    # primo passo, quindi non cambia con il numero di token.
    import torch
    stops = stop_ids(tokenizer, model)
    for row in rows:
        enc = tokenizer(chat_text(tokenizer, model_key, row["prompt"]), return_tensors="pt",
                        add_special_tokens=False).to(model.device)
        with torch.inference_mode():
            result = model.generate(**enc, do_sample=False, num_beams=1, min_new_tokens=0, min_length=0,
                                    max_new_tokens=max_new_tokens, use_cache=True,
                                    return_dict_in_generate=True, output_logits=True,
                                    pad_token_id=tokenizer.pad_token_id if tokenizer.pad_token_id is not None
                                    else tokenizer.eos_token_id)
        logits = result.logits[0].float()
        if not bool(torch.isfinite(logits).all()):
            raise RuntimeError("Logit non finiti al primo passo")
        p = logits.softmax(dim=-1)[0]
        probability = [float(p[ids[label]].sum()) for label in LABELS]
        new = result.sequences[0, enc["input_ids"].shape[1]:].tolist()
        text = tokenizer.decode(new, skip_special_tokens=True)
        stopped = any(t in stops for t in new)
        out = dict(row_id=row["row_id"], generated=text, parsed=parse_label(text), n_token_generati=len(new),
                   fine_naturale=bool(stopped),
                   prompt_tokens=int(enc["input_ids"].shape[1]), mass_total=sum(probability),
                   **{f"p_{k}": v for k, v in zip(LABELS, probability)})
        if keep_answer_ids:
            answer = []
            for t in new:                       # la risposta senza il token di fine turno
                if t in stops:
                    break
                answer.append(t)
            out["prompt_ids"] = enc["input_ids"][0].tolist()
            out["answer_ids"] = answer
        yield out


def load_causal_lm(model_key, model_id, revision, token):
    # Come il Test U2: bf16, attention eager, tutto sulla GPU 0, padding a sinistra.
    import torch
    import transformers
    from transformers import AutoModelForCausalLM, AutoTokenizer
    extra = {"fix_mistral_regex": True} if model_key == "mistral" else {}
    tokenizer = AutoTokenizer.from_pretrained(model_id, revision=revision, token=token, trust_remote_code=False, **extra)
    tokenizer.padding_side = "left"
    if tokenizer.pad_token_id is None:
        tokenizer.pad_token = tokenizer.eos_token
    major, minor = (int(x) for x in transformers.__version__.split(".")[:2])
    dtype_arg = "dtype" if (major, minor) >= (4, 56) else "torch_dtype"
    model = AutoModelForCausalLM.from_pretrained(model_id, revision=revision, token=token, device_map={"": 0},
                                                 attn_implementation="eager", trust_remote_code=False,
                                                 low_cpu_mem_usage=True, **{dtype_arg: torch.bfloat16}).eval()
    return model, tokenizer


def alignment(results, reference):
    # reference: [(pair_key, kind, valore)] da P−S del Test U2
    by_key = {(r["pair_key"], r["kind"]): r for r in results}
    errors = [abs(by_key[(k, kind)]["p_UNKNOWN"] - v) for k, kind, v in reference if (k, kind) in by_key]
    return dict(n=len(errors), errore_assoluto_max=max(errors) if errors else None)


def manifest(**extra):
    import platform
    info = dict(python=platform.python_version(), eseguibile=sys.executable)
    from importlib import metadata
    for name in ("torch", "transformers", "numpy", "peft", "datasets", "lm-polygraph", "accelerate", "tokenizers"):
        try:
            info[name] = metadata.version(name)
        except Exception:
            info[name] = None
    try:
        import torch
        info["gpu"] = torch.cuda.get_device_name(0) if torch.cuda.is_available() else None
    except Exception:
        pass
    info.update(extra)
    return info
```

```python
%%writefile /content/testC_run/script/tm_c_m1.py
# Test C, M1: risposte del modello (greedy, fino a 32 token), riferimento p(UNKNOWN) e stimatori di LM-Polygraph 0.7.0,
# per un modello. Gira in env_m1. Come tm_m1.py del Test M, senza i controlli positivi (HC0 si eredita dal Test M).
# Le risposte di questo lavoro sono quelle che si classificano (protocollo v3, §4). Uso: python tm_c_m1.py <config.json>
import gc
import os
import sys
from collections import defaultdict
from pathlib import Path

sys.path.insert(0, str(Path(__file__).parent))
from tm_comune import (checkpoint, clean, label_ids, load_causal_lm, manifest, predict_reference, prompt_ids,
                       read_json, say, write_json, alignment)

import numpy as np
import torch
from lm_polygraph import WhiteboxModel
from lm_polygraph.utils.dataset import Dataset
from lm_polygraph.utils.manager import UEManager
from lm_polygraph.utils.builder_enviroment_stat_calculator import BuilderEnvironmentStatCalculator
from lm_polygraph.utils.generation_parameters import GenerationParameters
from lm_polygraph.defaults.register_default_stat_calculators import register_default_stat_calculators
from lm_polygraph.estimators import (MaximumSequenceProbability, Perplexity, MeanTokenEntropy, SemanticEntropy, PTrue)

ESTIMATORS = {"MSP": MaximumSequenceProbability, "Perplexity": Perplexity, "MeanTokenEntropy": MeanTokenEntropy,
              "SemanticEntropy": SemanticEntropy, "PTrue": PTrue}
N_SAMPLES = 5            # protocollo §2.1: 5 campioni per Semantic Entropy
BATCH = 16


class FixedBatches(Dataset):
    # Dataset di LM-Polygraph con batch decisi da noi (solo l'ordine di iterazione cambia).
    def __init__(self, groups):
        super().__init__([x for g in groups for x in g], ["" for g in groups for _ in g], batch_size=BATCH)
        self.groups = groups

    def __iter__(self):
        for g in self.groups:
            yield (list(g), ["" for _ in g], None)

    def __len__(self):
        return len(self.groups)


def stat_calculators():
    # Calcolatori di default della libreria. Due sole scelte di configurazione:
    # niente attenzioni né stati interni nel passo greedy (non servono a questi stimatori),
    # e 5 campioni invece dei 10 di default, tramite un builder (punto di estensione previsto dalla libreria).
    calcs = register_default_stat_calculators("Whitebox", output_attentions=False, output_hidden_states=False)
    found = 0
    for sc in calcs:
        if sc.name == "SamplingGenerationCalculator":
            sc.builder = "tm_lmp_builders"
            sc.cfg["samples_n"] = N_SAMPLES
            found += 1
    if found != 1:
        raise RuntimeError("SamplingGenerationCalculator non trovato fra i calcolatori di default")
    return calcs


def groups_same_length(rows, lengths, padded):
    # B/C: batch con prompt della stessa lunghezza in token, quindi senza padding (risultati come con batch 1).
    # Controlli positivi: batch con padding a sinistra, come nell'uso standard della libreria.
    if padded:
        order = sorted(range(len(rows)), key=lambda i: lengths[i])
        return [order[i:i + BATCH] for i in range(0, len(order), BATCH)]
    by_len = defaultdict(list)
    for i, n in enumerate(lengths):
        by_len[n].append(i)
    out = []
    for n in sorted(by_len):
        idx = by_len[n]
        out.extend(idx[i:i + BATCH] for i in range(0, len(idx), BATCH))
    return out


def run_polygraph(wm, env, calcs, rows, lengths, max_new_tokens, padded, seed):
    groups = groups_same_length(rows, lengths, padded)
    texts = [[rows[i]["prompt"] for i in g] for g in groups]
    torch.manual_seed(seed)
    np.random.seed(seed % (2 ** 32))
    estimators = [cls() for cls in ESTIMATORS.values()]
    man = UEManager(FixedBatches(texts), wm, estimators, env, calcs, [], [], [], ignore_exceptions=False,
                    verbose=False, max_new_tokens=max_new_tokens, save_stats=["greedy_texts", "sample_texts"])
    man()
    order = [i for g in groups for i in g]
    out = [None] * len(rows)
    values = {name: man.estimations[("sequence", str(est))] for name, est in zip(ESTIMATORS, estimators)}
    for name, v in values.items():
        if len(v) != len(rows):
            raise RuntimeError(f"{name}: {len(v)} stime per {len(rows)} prompt")
    greedy = man.stats["greedy_texts"]
    samples = man.stats["sample_texts"]
    for pos, i in enumerate(order):
        out[i] = dict(row_id=rows[i]["row_id"], greedy_polygraph=greedy[pos], campioni=list(samples[pos]),
                      **{f"u_{name}": clean(values[name][pos]) for name in ESTIMATORS})
    return out


def configure_natural_generation(wm, model, tokenizer):
    """Toglie il minimo forzato della libreria; conserva gli EOS nativi e il limite del chiamante.

    Vale per greedy, campionamento e PTrue. Non si cambia la temperatura dei campioni.
    La differenza da LM-Polygraph 0.7.0 è dichiarata nel manifest.
    """
    native_config = model.generation_config
    eos = getattr(native_config, "eos_token_id", None)
    if eos is None:
        eos = tokenizer.eos_token_id
    original = wm.generate
    repetition = getattr(native_config, "repetition_penalty", 1.0) or 1.0

    def generate(**kwargs):
        kwargs["min_new_tokens"] = 0
        kwargs["min_length"] = 0
        if eos is not None:
            kwargs["eos_token_id"] = eos
        kwargs["pad_token_id"] = (tokenizer.pad_token_id if tokenizer.pad_token_id is not None
                                   else tokenizer.eos_token_id)
        # Il riferimento è greedy a un fascio; i parametri dei campioni restano quelli della libreria.
        sampling = kwargs.get("do_sample", wm.generation_parameters.do_sample)
        if not sampling:
            kwargs["do_sample"] = False
            kwargs["num_beams"] = 1
            kwargs["repetition_penalty"] = repetition
        return original(**kwargs)

    wm.generate = generate
    return dict(scelta="naturale", min_new_tokens=0, min_length=0, eos_token_id=eos,
                pad_token_id=(tokenizer.pad_token_id if tokenizer.pad_token_id is not None else tokenizer.eos_token_id),
                repetition_penalty_greedy=repetition,
                minimo_libreria_originale=2, applicata_anche_alle_generazioni_ausiliarie=True,
                stimatori="LM-Polygraph 0.7.0 originali")


def main(cfg_path):
    cfg = read_json(cfg_path)
    if cfg.get("m1_generazione") != "naturale":
        raise ValueError("La v2 richiede m1_generazione=naturale, fissato prima della prova")
    key, out = cfg["modello"], Path(cfg["uscita"])
    out.mkdir(parents=True, exist_ok=True)
    say(f"M1 {key}: carico {cfg['id']} ({cfg['sha'][:8]})")
    model, tokenizer = load_causal_lm(key, cfg["id"], cfg["sha"], os.environ.get("HF_TOKEN"))
    ids = label_ids(tokenizer)
    if ids != cfg["label_ids"]:
        raise RuntimeError(f"{key}: token delle etichette diversi dal Test U")

    rows = read_json(cfg["righe"])
    ref_rows = read_json(cfg["righe_allineamento"])
    # 1. Riferimento p(UNKNOWN) e risposta del modello: batch 1, greedy, fino a 32 token (protocollo v3, §4)
    check = list(predict_reference(model, tokenizer, key, ref_rows, ids))
    al = alignment([dict(r, **c) for r, c in zip(ref_rows, check)], cfg["riferimento_PN"])
    say(f"  allineamento P−S con il Test U2: max |delta| = {al['errore_assoluto_max']:.3g} su {al['n']} valori")
    n_tok = int(cfg["token_risposta"])
    reference = checkpoint(out, "risposte", rows,
                           lambda chunk, i: predict_reference(model, tokenizer, key, chunk, ids, max_new_tokens=n_tok),
                           size=128)

    # 2. LM-Polygraph. Stessi token del riferimento: il chat template di Qwen si chiama con enable_thinking=False.
    if key == "qwen":
        original = tokenizer.apply_chat_template

        def no_thinking(*args, **kwargs):
            kwargs.setdefault("enable_thinking", False)
            return original(*args, **kwargs)
        tokenizer.apply_chat_template = no_thinking
    tokenizer.padding_side = "left"
    wm = WhiteboxModel(model, tokenizer, model_path=cfg["id"], model_type="CausalLM",
                       generation_parameters=GenerationParameters(), instruct=True)
    generation_policy = configure_natural_generation(wm, model, tokenizer)
    say(f"  generazione M1: stop naturale, min_new_tokens=0, massimo {n_tok} token")
    mismatch = 0
    lengths = {}
    for r in rows:
        mine = prompt_ids(tokenizer, key, r["prompt"])
        theirs = wm.tokenize([r["prompt"]])["input_ids"][0].tolist()
        mismatch += mine != theirs
        lengths[r["row_id"]] = len(theirs)
    say(f"  token di LM-Polygraph diversi dal riferimento: {mismatch} su {len(rows)}")
    if mismatch:
        raise RuntimeError("LM-Polygraph non vede gli stessi token del riferimento: M1 non si esegue")
    env = BuilderEnvironmentStatCalculator(wm)
    calcs = stat_calculators()
    seed = int(cfg["seme"])
    polygraph = checkpoint(out, "polygraph", rows,
                           lambda chunk, i: run_polygraph(wm, env, calcs, chunk, [lengths[r["row_id"]] for r in chunk],
                                                          n_tok, False, seed + i), size=64)
    uguali = sum(p["greedy_polygraph"].strip() == r["generated"].strip() for p, r in zip(polygraph, reference))
    say(f"  risposta greedy di LM-Polygraph uguale alla risposta del riferimento: {uguali} su {len(rows)}")

    result = dict(manifest(metodo="M1", modello=key, id=cfg["id"], sha=cfg["sha"], campioni=N_SAMPLES, batch=BATCH,
                           lm_polygraph_campioni_default=10),
                  generazione_M1=generation_policy, hc0_configurazione="ereditato dal Test M, minimo 2: applicabilita da riverificare prima del completo",
                  allineamento=al, token_diversi=mismatch, token_risposta=n_tok, riferimento=reference,
                  polygraph=polygraph, greedy_uguale_al_riferimento=uguali)
    write_json(out / f"M1_{key}.json", result)
    say(f"M1 {key}: completato")
    del model, wm
    gc.collect()
    torch.cuda.empty_cache()


if __name__ == "__main__":
    main(sys.argv[1])
```

```python
%%writefile /content/testC_run/script/tm_ref.py
# Riferimento p(UNKNOWN) di un modello in un ambiente qualsiasi (Test C: env_sirin, per i ΔW di M2 e M3).
# p(UNKNOWN) dipende solo dal primo passo: qui basta generare token_risposta token (cfg).
# Uso: python tm_ref.py <config.json>
import gc
import os
import sys
from pathlib import Path

sys.path.insert(0, str(Path(__file__).parent))
from tm_comune import alignment, checkpoint, label_ids, load_causal_lm, manifest, predict_reference, read_json, say, write_json

import torch


def main(cfg_path):
    cfg = read_json(cfg_path)
    key, out = cfg["modello"], Path(cfg["uscita"])
    out.mkdir(parents=True, exist_ok=True)
    say(f"Riferimento {cfg['ambiente']} {key}: carico {cfg['id']} ({cfg['sha'][:8]})")
    model, tokenizer = load_causal_lm(key, cfg["id"], cfg["sha"], os.environ.get("HF_TOKEN"))
    ids = label_ids(tokenizer)
    if ids != cfg["label_ids"]:
        raise RuntimeError(f"{key}: token delle etichette diversi dal Test U")
    ref_rows = read_json(cfg["righe_allineamento"])
    check = list(predict_reference(model, tokenizer, key, ref_rows, ids))
    al = alignment([dict(r, **c) for r, c in zip(ref_rows, check)], cfg["riferimento_PN"])
    say(f"  allineamento P−S con il Test U2: max |delta| = {al['errore_assoluto_max']:.3g} su {al['n']} valori")
    rows = read_json(cfg["righe"])
    reference = checkpoint(out, f"riferimento_{cfg['ambiente']}", rows,
                           lambda chunk, i: predict_reference(model, tokenizer, key, chunk, ids,
                                                             max_new_tokens=int(cfg["token_risposta"])), size=128)
    write_json(out / f"REF_{cfg['ambiente']}_{key}.json",
               dict(manifest(metodo="riferimento", ambiente=cfg["ambiente"], modello=key, id=cfg["id"], sha=cfg["sha"]),
                    allineamento=al, riferimento=reference))
    say(f"Riferimento {cfg['ambiente']} {key}: completato")
    del model
    gc.collect()
    torch.cuda.empty_cache()


if __name__ == "__main__":
    main(sys.argv[1])
```

```python
%%writefile /content/testC_run/script/tm_m2_comune.py
# Funzioni comuni agli script di M2 (env_sirin): testo d'ingresso, resa con il chat template di SIRIN, lunghezze,
# confine dell'assistente e sottoinsiemi annidati dei dati di addestramento.
import re
import sys
from pathlib import Path

sys.path.insert(0, str(Path(__file__).parent))

MAX_TOKEN = 512


def inspection_prompt(context, question):
    # come M3 e come _inspection_prompt dell'interfaccia di SIRIN (nota M2 v2, §2.5)
    return f'Context:\n{context.strip()}\n\nQuestion:\n{question.strip()}'


def sample_of(row):
    # campione di SIRIN: turno utente con il testo d'ingresso, turno dell'assistente vuoto, nessun prompt di sistema
    return [{"role": "user", "content": inspection_prompt(row["contesto"], row["domanda"])},
            {"role": "assistant", "content": ""}]


class Renderer:
    """Resa del campione come la fa HfModelAdapter di SIRIN (stesso tokenizer, stesso _preprocess_input).

    Il tokenizer si carica con la stessa chiamata di _load_model_and_tokenizer (AutoTokenizer, revisione fissata),
    senza caricare i pesi del modello."""

    def __init__(self, model_id, revision, tokenizer=None, adapter=None):
        from sirin.inference.adapters import HfModelAdapter
        from sirin.models.inference import HFConfig
        from transformers import AutoTokenizer
        if adapter is None:
            adapter = HfModelAdapter(HFConfig(model_path=model_id, revision=revision, device="cpu"))
            tok = tokenizer or AutoTokenizer.from_pretrained(model_id, revision=revision)
            if adapter.config.max_length is not None:
                tok.max_length = min(adapter.config.max_length, tok.model_max_length)
            tok.padding_side = adapter.config.padding_side or "right"
            adapter.tokenizer = tok
        self.adapter = adapter
        self.tok = adapter.tokenizer
        # Il localizzatore di SIRIN (HfTokenLocator.locate) decodifica i token del prompt e li ritokenizza per avere gli
        # offset. Con clean_up_tokenization_spaces=True (default di alcuni tokenizer) la decodifica toglie gli spazi
        # prima di «, . ? ! 's n't…», la ritokenizzazione dà meno token e il localizzatore si ferma (prova del 11
        # novembre: «200 token ids but 198 offsets»). La pulizia agisce solo sulla decodifica: la si spegne sull'istanza
        # del tokenizer, così la decodifica è il testo esatto. La tokenizzazione, e quindi gli stati, non cambiano.
        self.pulizia_spazi_originale = bool(getattr(self.tok, "clean_up_tokenization_spaces", False))
        self.tok.clean_up_tokenization_spaces = False

    def render(self, sample):
        return self.adapter._preprocess_input([sample])[0]

    def ids(self, text):
        # come generate_hiddens: add_special_tokens=False sul testo già reso
        return self.tok(text, add_special_tokens=False)["input_ids"]

    def full_length(self, row):
        return len(self.ids(self.render(sample_of(row))))

    def boundary(self, row):
        """Controlli del confine dell'assistente per un campione.

        n_full: token dell'intero prompt (quelli che entrano nel passaggio in avanti, senza padding né troncamento);
        n_user: token della resa del solo turno utente; ans_start: primo token della parte dell'assistente secondo il
        localizzatore di SIRIN (HfTokenLocator, locate_answer_start), cioè il confine usato da side=LEFT;
        prefix_ok: la resa completa comincia con quella del solo turno utente."""
        from sirin.inference.token_locators import HfTokenLocator
        from sirin.models.inference import TokenLocatorConfig
        sample = sample_of(row)
        full = self.render(sample)
        user = self.adapter._preprocess_input([sample[:1]])[0]
        full_ids = self.ids(full)
        answer = full[len(user):]                    # come locate_tokens di HfModelAdapter
        roundtrip = self.ids(self.tok.decode(full_ids, skip_special_tokens=False)) == full_ids
        try:
            loc = HfTokenLocator(TokenLocatorConfig(locate_answer_start=True)).locate(
                tokens=full_ids, tokenizer=self.tok, answer_text=answer)
            ans_start, err = int(loc["answer_start"]), None
        except ValueError as e:                     # il localizzatore di SIRIN non trova il confine: errore registrato
            ans_start, err = -1, str(e)[:300]
        return dict(n_full=len(full_ids), n_user=len(self.ids(user)), ans_start=ans_start,
                    prefix_ok=full.startswith(user), roundtrip_ok=roundtrip, errore_localizzatore=err, suffix=answer)


def word_ends(text):
    # fine (carattere escluso) di ogni parola, dove una parola è una sequenza di caratteri non bianchi
    return [m.end() for m in re.finditer(r"\S+", text)]


def nested(rows, n_total):
    """Sottoinsieme annidato: le prime n_total/2 righe di ciascuna classe nell'ordine fissato dalla selezione,
    poi l'ordine di addestramento fissato (campo 'ordine'). Le dimensioni piccole sono contenute nelle grandi."""
    assert n_total % 2 == 0, n_total
    half = n_total // 2
    out = []
    for label in (0, 1):
        part = [r for r in rows if r["label"] == label][:half]
        if len(part) != half:
            raise RuntimeError(f"servono {half} righe della classe {label}, ce ne sono {len(part)}")
        out += part
    return sorted(out, key=lambda r: r["ordine"])
```

```python
%%writefile /content/testC_run/script/tm_c_m2.py
# Test C, M2: sonde di answerability di SIRIN salvate dal completo del notebook B (M2 originale e M2-senzaℓ2), applicate
# alle righe del Test C senza riaddestrarle (protocollo v3, §5). Gira in env_sirin. Uso: python tm_c_m2.py <config.json>
#
# Stessa estrazione del notebook B (tm_m2.py): adattatore di SIRIN in bf16, sdpa, chat template, batch 1, strato
# centrale, lato LEFT con confine dell'assistente controllato, e stesso detector (SequenceLinearProbingDetector, stessa
# configurazione e stesso seme). I pesi della sonda (state_dict completo, query compresa) e lo scaler si caricano dai file
# di B. Controllo del ricaricamento: le prime domande di verifica di B (SQuAD) si ripuntano qui e si confrontano con i
# punteggi salvati da B; lo scarto massimo deve stare entro la tolleranza (1e-6), altrimenti la variante non si usa.
import gc
import sys
import time
from pathlib import Path

sys.path.insert(0, str(Path(__file__).parent))
import numpy as np
import torch

from tm_comune import clean, manifest, read_json, say, write_json
from tm_m2_comune import MAX_TOKEN, Renderer, sample_of

VARIANTI = (("originale", "sonda"), ("senza_l2", "sonda_senzaL2"))


def build_classifier(det):
    # stesso costruttore di SequenceLinearProbingDetector.setup_model (come in tm_m2.py del notebook B)
    from sirin.classification import LinearClassifier
    c = det.config
    return LinearClassifier(embedding_dim=c.embedding_dim, num_features=c.num_features,
                            attention_pooling=c.attention_pooling, ensemble=c.ensemble,
                            num_classes=c.num_classification_heads, use_contrastive=c.use_contrastive,
                            projection_dim=c.projection_dim, contrastive_layers=c.contrastive_layers,
                            projection_hidden_dim=getattr(c, "projection_hidden_dim", None),
                            projection_num_layers=getattr(c, "projection_num_layers", 2),
                            use_projection_dropout=getattr(c, "use_projection_dropout", False),
                            projection_dropout=getattr(c, "projection_dropout", 0.1))


def main(cfg_path):
    cfg = read_json(cfg_path)
    key, out = cfg["modello"], Path(cfg["uscita"])
    out.mkdir(parents=True, exist_ok=True)
    from loguru import logger
    logger.remove()
    logger.add(sys.stderr, level="INFO")
    t_start = time.time()

    from sirin.detection.probing.detectors.sequence.linear import SequenceLinearProbingDetector
    from sirin.detection.probing.detectors.utils.detection import check_features_for_nan
    from sirin.detection.probing.preprocessing import FeaturePreprocessor
    from sirin.detection.processors.hidden import HiddensProcessor
    from sirin.definitions import SideType
    from sirin.inference.adapters import HfModelAdapter
    from sirin.models.detection import HiddensProcessorConfig, ProbingDetectorConfig
    from sirin.models.inference import HFConfig, TokenLocatorConfig

    # --- lavoro M2 del notebook B: configurazione, rapporto e file delle sonde
    cart_b = Path(cfg["cartella_B"])
    cfg_b = read_json(cart_b / "config.json")
    rep_b = read_json(cart_b / f"M2_{key}.json")
    if cfg_b["id"] != cfg["id"] or cfg_b["sha"] != cfg["sha"]:
        raise RuntimeError(f"{key}: modello del notebook B {cfg_b['id']}@{cfg_b['sha']} diverso da {cfg['id']}@{cfg['sha']}")
    if rep_b.get("sirin_commit") != cfg["sirin_commit"]:
        raise RuntimeError(f"{key}: commit di SIRIN del notebook B {rep_b.get('sirin_commit')} diverso da {cfg['sirin_commit']}")
    viste_b = {"originale": rep_b, "senza_l2": dict(rep_b, **rep_b.get("senza_l2", {}))}

    say(f"M2 {key}: carico {cfg['id']} ({(cfg['sha'] or '')[:8]})")
    adapter = HfModelAdapter(HFConfig(model_path=cfg["id"], revision=cfg["sha"], model_dtype="bf16",
                                      attn_implementation="sdpa", use_chat_template=True, batch_size=1,
                                      device=cfg.get("dispositivo", "cuda")))
    adapter.load()
    n_layers = int(adapter.model.config.num_hidden_layers)
    layer = n_layers // 2
    if n_layers != cfg_b["strati_attesi"] or layer != cfg_b["strato"] or layer != rep_b["strato"]:
        raise RuntimeError(f"{key}: {n_layers} strati, strato {layer}; nel notebook B strato {rep_b['strato']}")
    renderer = Renderer(cfg["id"], cfg["sha"], adapter=adapter)

    def expected(rows, what):
        exp = []
        for r in rows:
            b = renderer.boundary(r)
            if (b["n_full"] > MAX_TOKEN or not b["prefix_ok"] or not b["roundtrip_ok"] or b["errore_localizzatore"]
                    or b["ans_start"] != b["n_user"]):
                raise RuntimeError(f"{what} {r['row_id']}: controllo del confine non superato {b}")
            exp.append(b["n_user"])
        return exp

    proc = HiddensProcessor(
        HiddensProcessorConfig(layers=[layer], side=SideType.LEFT, pooling_type="none", separate=True,
                               cache_features=False, padding=False, truncation=False,
                               token_locator_config=TokenLocatorConfig(locate_answer_start=True)),
        extractor=adapter)
    seen = []
    orig_extract = proc.extract_features

    def extract_features(*a, **k):
        feats, idx = orig_extract(*a, **k)
        seen.extend([list(map(int, i)) for i in idx])
        return feats, idx
    proc.extract_features = extract_features

    def check_seen(exp, what):
        if len(seen) != len(exp):
            raise RuntimeError(f"{what}: {len(seen)} campioni estratti invece di {len(exp)}")
        for i, (got, n) in enumerate(zip(seen, exp)):
            if got != list(range(n)):
                raise RuntimeError(f"{what}, campione {i}: indici LEFT {got[:3]}…{got[-3:]} (n={len(got)}) invece di 0…{n - 1}")
        seen.clear()

    # stesso detector del notebook B (configurazione, seme); model_save_path in una cartella di questo lavoro, mai usata
    det_cfg = ProbingDetectorConfig(attention_pooling=[True], num_classification_heads=1, use_contrastive=False,
                                    max_length=MAX_TOKEN, seed=cfg_b["seme_libreria"], device=cfg.get("dispositivo", "cuda"),
                                    model_save_path=str(out / "non_usata"), batch_size=1)
    det = SequenceLinearProbingDetector(config=det_cfg, feature_processor=proc)
    seen.clear()                                 # campione di prova interno della libreria (_cold_run)

    sonde, scaler, info_sonde = {}, {}, {}
    for nome, cartella in VARIANTI:
        ck_path = cart_b / cartella / "m2_state.pt"
        if not ck_path.is_file():
            info_sonde[nome] = dict(errore=f"file mancante: {ck_path}")
            continue
        torch.manual_seed(12345)                 # inizializzazione qualsiasi: i pesi arrivano tutti dal file
        m = build_classifier(det)
        ck = torch.load(ck_path, map_location="cpu", weights_only=False)
        if ck.get("variante") != nome:
            raise RuntimeError(f"{ck_path}: variante {ck.get('variante')} invece di {nome}")
        m.load_state_dict(ck["state_dict"], strict=True)
        m.to(det.device).eval()
        sonde[nome] = m
        scaler[nome] = FeaturePreprocessor.load(str(cart_b / cartella / "compressor"))
        info_sonde[nome] = dict(file=str(ck_path), epoca=ck.get("epoca"),
                                somma_parametri=float(sum(v.double().sum() for v in ck["state_dict"].values())),
                                ricaricamento_B=viste_b[nome].get("ricaricamento"))
        say(f"  sonda {nome}: epoca {ck.get('epoca')}, da {ck_path}")

    def score(rows, what):
        exp = expected(rows, what)
        res = {nome: [] for nome in sonde}
        for start in range(0, len(rows), 16):    # stessi blocchi da 16 del notebook B
            chunk = rows[start:start + 16]
            raw = None
            for r in chunk:
                f, _ = det.feature_processor([sample_of(r)])
                check_features_for_nan(f, [r])
                raw = f if raw is None else [a + b for a, b in zip(raw, f)]
            for nome, model in sonde.items():
                old = det.compressor
                det.compressor = scaler[nome]
                feats, masks = det._process_sequence_features(raw)
                det.compressor = old
                with torch.no_grad():
                    logits = model([x.to(det.device) for x in feats], [m.to(det.device) for m in masks])
                probs, _ = det._logits_to_probs_preds(logits)
                res[nome].extend(np.atleast_1d(probs).tolist())
            if start == 0 or (start // 16 + 1) % 25 == 0:
                say(f"  punteggi {what} {min(start + 16, len(rows))}/{len(rows)}")
        check_seen(exp, what)
        return {nome: [dict(row_id=r["row_id"], p_non_risp=clean(x)) for r, x in zip(rows, v)] for nome, v in res.items()}

    # --- controllo del ricaricamento sulle prime domande di verifica di B
    dati_b = read_json(cfg_b["dati"])
    n_ctrl = int(cfg["n_controllo"])
    ctrl_rows = dati_b["verifica"][:n_ctrl]
    ctrl = score(ctrl_rows, "controllo")
    tol = float(cfg["tolleranza_ricaricamento"])
    controllo = {}
    for nome in sonde:
        saved = {r["row_id"]: r["p_non_risp"] for r in viste_b[nome]["verifica"]}
        diffs = [abs(r["p_non_risp"] - saved[r["row_id"]]) for r in ctrl[nome]
                 if r["p_non_risp"] is not None and saved.get(r["row_id"]) is not None]
        mx = max(diffs) if len(diffs) == len(ctrl_rows) else None
        controllo[nome] = dict(domande=len(ctrl_rows), confrontate=len(diffs), scarto_max=mx, tolleranza=tol,
                               superato=mx is not None and mx <= tol)
        say(f"  ricaricamento {nome}: scarto max {mx} su {len(diffs)} domande di verifica (tolleranza {tol})")

    # --- punteggi sulle righe del Test C
    rows = read_json(cfg["righe"])
    t0 = time.time()
    punteggi = score(rows, "righe")
    say(f"  righe: {len(rows)} in {time.time() - t0:.0f} s")

    report = dict(manifest(metodo="M2", modello=key, id=cfg["id"], sha=cfg["sha"], sirin_commit=cfg["sirin_commit"]),
                  strato=layer, strati=n_layers, pulizia_spazi_originale=renderer.pulizia_spazi_originale,
                  cartella_B=str(cart_b), sonde=info_sonde, controllo_ricaricamento=controllo,
                  righe={nome: punteggi[nome] for nome in sonde}, secondi_totali=time.time() - t_start)
    write_json(out / f"M2_{key}.json", report)
    if not controllo.get("originale", {}).get("superato"):
        say(f"M2 {key}: controllo del ricaricamento della sonda originale NON superato")
    say(f"M2 {key}: completato in {report['secondi_totali'] / 60:.1f} min")
    del adapter
    gc.collect()
    torch.cuda.empty_cache()


if __name__ == "__main__":
    main(sys.argv[1])
```

```python
%%writefile /content/testC_run/script/tm_m3.py
# Test C, M3: giudice zero-shot di answerability di SIRIN (Qwen3-4B), come nel Test M, sulle righe del Test C (un
# punteggio per prompt, uguale per tutti i modelli). HC0 si eredita dal Test M: niente SQuAD. Gira in env_sirin.
# Uso: python tm_m3.py <config.json>
import ast
import sys
from pathlib import Path

sys.path.insert(0, str(Path(__file__).parent))
from tm_comune import checkpoint, clean, manifest, read_json, say, write_json


def sirin_constant(path, name):
    # legge la costante dal sorgente di SIRIN senza importare l'interfaccia grafica (streamlit)
    tree = ast.parse(Path(path).read_text())
    for node in tree.body:
        if isinstance(node, ast.Assign) and any(getattr(t, "id", None) == name for t in node.targets):
            return ast.literal_eval(node.value)
    raise RuntimeError(f"{name} non trovato in {path}")


def check_inspection_prompt(path):
    # La funzione _inspection_prompt dell'interfaccia di SIRIN costruisce «Context: … Question: …».
    src = Path(path).read_text()
    expected = "return f'Context:\\n{context.strip()}\\n\\nQuestion:\\n{question.strip()}'"
    if expected not in src:
        raise RuntimeError("_inspection_prompt di SIRIN è diverso da quello del protocollo")


def inspection_prompt(context, question):
    return f'Context:\n{context.strip()}\n\nQuestion:\n{question.strip()}'


def main(cfg_path):
    cfg = read_json(cfg_path)
    out = Path(cfg["uscita"])
    out.mkdir(parents=True, exist_ok=True)
    root = Path(cfg["sirin_repo"])
    prompt = sirin_constant(root / "sirin/ui/presets.py", "_ANSWERABILITY_JUDGE_PROMPT")
    if prompt != cfg["prompt_atteso"]:
        raise RuntimeError("Il prompt di answerability di SIRIN non coincide con quello del protocollo (§2.2)")
    check_inspection_prompt(root / "sirin/ui/streamlit_app.py")

    from sirin.detection.judging.judges.sequence.decoder import SequenceDecoderJudge
    from sirin.inference.adapters import HfModelAdapter
    from sirin.models.detection import HfJudgeConfig
    from sirin.models.inference import HFConfig

    say(f"M3: carico il giudice {cfg['giudice']} ({(cfg['giudice_sha'] or '')[:8]})")
    adapter = HfModelAdapter(HFConfig(model_path=cfg["giudice"], revision=cfg["giudice_sha"], device=cfg.get("dispositivo", "cuda"),
                                      batch_size=1))
    # Configurazione del preset «Judge — API Answerability (zero-shot)» di SIRIN, con il giudice locale:
    # stesso user_prompt, dialogue_format='{question}', system_prompt di default della libreria.
    # num_classification_heads=2: probabilità di «1» normalizzata su {«0», «1»}; max_new_tokens=1: si legge
    # solo il primo token (il protocollo a un token di SIRIN, verdict_max_tokens=1).
    judge = SequenceDecoderJudge(config=HfJudgeConfig(user_prompt=prompt, dialogue_format="{question}",
                                                      num_classification_heads=2, max_new_tokens=1,
                                                      device=cfg.get("dispositivo", "cuda")),
                                 model_adapter=adapter)
    # Protocollo §2.1: Qwen3-4B con enable_thinking=False. Il giudice locale di SIRIN unisce il dialogo senza prompt di
    # generazione e aggiunge solo «<|im_start|>assistant\n», cioè il modo con ragionamento: il primo token atteso è
    # allora «<think>» e le cifre 0/1 non dicono nulla (prova GPU: p≈0.7 su tutte le domande SQuAD). Si usa il prompt di
    # generazione del chat template con enable_thinking=False, come fa il preset di SIRIN per il giudice via API.
    tok = adapter.tokenizer
    probe_msg = [{"role": "user", "content": "x"}]
    base = tok.apply_chat_template(probe_msg, tokenize=False, add_generation_prompt=False)
    gen = tok.apply_chat_template(probe_msg, tokenize=False, add_generation_prompt=True, enable_thinking=False)
    if not gen.startswith(base) or not gen[len(base):].startswith(judge.assistant_prefix):
        raise RuntimeError(f"Prompt di generazione inatteso: {gen[len(base):]!r}")
    prefisso_libreria = judge.assistant_prefix
    judge.assistant_prefix = gen[len(base):]
    say(f"  token delle classi: {judge.class_token_ids}, prefisso dell'assistente: {judge.assistant_prefix!r} "
        f"(libreria: {prefisso_libreria!r})")

    def score(chunk, i):
        res = []
        for r in chunk:                    # un campione alla volta: nessun padding (il tokenizer pad a destra)
            sample = [{"role": "user", "content": inspection_prompt(r["contesto"], r["domanda"])},
                      {"role": "assistant", "content": ""}]
            probs, preds, _ = judge.detect([sample])
            res.append(dict(row_id=r["row_id"], p_non_risp=clean(probs[0]), pred=int(preds[0])))
        return res

    results = {}
    rows = read_json(cfg["righe"])
    unici = list({(r["contesto"], r["domanda"]): r for r in rows}.values())   # un punteggio per contesto e domanda
    results["righe"] = checkpoint(out, "M3_righe", unici, score, size=128)
    import transformers
    write_json(out / "M3.json", dict(manifest(metodo="M3", giudice=cfg["giudice"], sha=cfg["giudice_sha"],
                                              transformers_giudice=transformers.__version__,
                                              sirin_commit=cfg["sirin_commit"],
                                              system_prompt=judge.config.system_prompt,
                                              prefisso=judge.assistant_prefix, prefisso_libreria=prefisso_libreria,
                                              classi=judge.class_token_ids),
                                     righe=results["righe"]))
    say("M3: completato")


if __name__ == "__main__":
    main(sys.argv[1])
```

```python
%%writefile /content/testC_run/script/tm_c_m4.py
# Test C, M4: sonde rilasciate da Obeso et al. 2025 (lineare e LoRA-KL) su Llama-3.1-8B-Instruct, più il riferimento
# p(UNKNOWN) di env_obeso. Come tm_m4.py del Test M, con la risposta generata fino a 32 token e senza il controllo
# positivo (HC0 si eredita dal Test M). Gira in env_obeso, con la cartella del repository degli autori come directory
# corrente. Uso: python tm_c_m4.py <config.json>
import gc
import os
import sys
from pathlib import Path

sys.path.insert(0, str(Path(__file__).parent))
from tm_comune import (alignment, checkpoint, clean, label_ids, load_causal_lm, manifest, predict_reference,
                       read_json, say, write_json)

import torch


def probe_scores(probe, rows, gen):
    # Forward su prompt + risposta generata da Llama base; massimo della probabilità della sonda sui token della risposta.
    out = []
    for r in rows:
        g = gen[r["row_id"]]
        ids = g["prompt_ids"] + g["answer_ids"]
        n_prompt, n_answer = len(g["prompt_ids"]), len(g["answer_ids"])
        if n_answer == 0:
            out.append(dict(row_id=r["row_id"], sonda_max=None, sonda_token=[]))
            continue
        x = torch.tensor([ids], device=probe.device)
        with torch.inference_mode():
            o = probe(input_ids=x, attention_mask=torch.ones_like(x))
        p = torch.sigmoid(o["probe_logits"].squeeze(-1)[0].float())[n_prompt:n_prompt + n_answer]
        out.append(dict(row_id=r["row_id"], sonda_max=clean(p.max()), sonda_token=[clean(v) for v in p.tolist()]))
    return out


def main(cfg_path):
    cfg = read_json(cfg_path)
    out = Path(cfg["uscita"])
    out.mkdir(parents=True, exist_ok=True)
    sys.path.insert(0, cfg["obeso_repo"])
    from probe.config import ProbeConfig
    from probe.value_head_probe import setup_probe
    from utils.probe_loader import LOCAL_PROBES_DIR

    for probe_id in cfg["sonde"]:
        if not (Path(LOCAL_PROBES_DIR) / probe_id / "probe_head.bin").exists():
            raise RuntimeError(f"Sonda {probe_id} non scaricata in {LOCAL_PROBES_DIR}")

    say(f"M4: carico {cfg['id']} ({cfg['sha'][:8]})")
    model, tokenizer = load_causal_lm("llama", cfg["id"], cfg["sha"], os.environ.get("HF_TOKEN"))
    ids = label_ids(tokenizer)
    if ids != cfg["label_ids"]:
        raise RuntimeError("llama: token delle etichette diversi dal Test U")
    rows = read_json(cfg["righe"])
    ref_rows = read_json(cfg["righe_allineamento"])
    check = list(predict_reference(model, tokenizer, "llama", ref_rows, ids))
    al = alignment([dict(r, **c) for r, c in zip(ref_rows, check)], cfg["riferimento_PN"])
    say(f"  allineamento P−S con il Test U2: max |delta| = {al['errore_assoluto_max']:.3g} su {al['n']} valori")
    # 1. Risposte di Llama base e p(UNKNOWN): una sola generazione, valutata poi da entrambe le sonde
    gen_rows = checkpoint(out, "M4_generazione", rows,
                          lambda chunk, i: predict_reference(model, tokenizer, "llama", chunk, ids, keep_answer_ids=True,
                                                             max_new_tokens=int(cfg["token_risposta"])),
                          size=128)
    gen = {g["row_id"]: g for g in gen_rows}

    # 2. Sonde, nell'ordine dato: prima la lineare (modello base), poi la LoRA (l'adattatore modifica il modello)
    scores = {}
    for probe_id in cfg["sonde"]:
        pc = ProbeConfig(probe_id=probe_id, model_name=cfg["id"], layer=30, load_from="disk")
        model_p, probe = setup_probe(model, pc)
        say(f"  sonda {probe_id}: strato {probe.layer_idx}, LoRA {'sì' if model_p is not model else 'no'}")
        scores[probe_id] = checkpoint(out, f"M4_{probe_id}", rows, lambda chunk, i: probe_scores(probe, chunk, gen),
                                      size=128)
        model = model_p
        del probe
    del model, model_p
    gc.collect()
    torch.cuda.empty_cache()
    # gli id del prompt servono solo qui: nel file finale restano gli id della risposta
    gen_out = [{k: v for k, v in g.items() if k != "prompt_ids"} for g in gen_rows]
    result = dict(manifest(metodo="M4", id=cfg["id"], sha=cfg["sha"], sonde=cfg["sonde"],
                           revisione_sonde=cfg["sonde_sha"], obeso_commit=cfg["obeso_commit"]),
                  allineamento=al, token_risposta=int(cfg["token_risposta"]), generazione=gen_out, sonde=scores)
    write_json(out / "M4.json", result)
    say("M4: completato")


if __name__ == "__main__":
    main(sys.argv[1])
```

```python
%%writefile /content/testC_run/script/tc_classifica.py
# Test C: classificazione delle risposte (protocollo v3, §4). Solo libreria standard: gira nel Python di Colab.
# Le regole sono scritte prima dei dati; si possono correggere solo sui 10 passaggi della prova (§8), poi si fissano.
#
# Ordine delle regole (§4): prima l'astensione, poi le alternative e le attribuzioni (con «entrambi affermati»), poi la
# qualificazione, poi gli altri casi. Una risposta come «1821 or maybe 1823» è quindi «alternative».
import re
import string

VERSIONE_REGOLE = "C-classifica-2"

CATEGORIE = ("corretta", "errata", "astensione", "alternative", "attribuite", "entrambi affermati", "qualificata",
             "un valore del contesto", "valore assente dal contesto", "altro")
SUPPORTATE_C = ("alternative", "attribuite")
NON_SUPPORTATE_C = ("entrambi affermati", "un valore del contesto", "valore assente dal contesto")
SUPPORTATE_B = ("corretta",)
NON_SUPPORTATE_B = ("errata", "valore assente dal contesto")
ESCLUSE = ("astensione", "qualificata", "altro")

# formule di non risposta (elenco fisso; si confrontano sul testo in minuscolo)
ASTENSIONE = re.compile(
    r"\bunknown\b|\bnot (?:explicitly )?(?:mentioned|stated|specified|provided|given|named|known|clear|determinable|"
    r"indicated|said)\b|\bno (?:information|answer|mention)\b|\b(?:does|do|did)(?: not|n't) (?:say|mention|specify|"
    r"state|provide|indicate|tell|give)\b|\bcannot be (?:determined|known|answered)\b|\bcan't be (?:determined|known)\b|"
    r"\b(?:cannot|can't|unable to) (?:determine|answer|tell|know)\b|\bunclear\b|\bi (?:do not|don't) know\b|"
    r"\bnot enough information\b|\binsufficient information\b|\bunanswerable\b|\bimpossible to (?:determine|know|tell)\b")
# riserve esplicite (elenco fisso)
MARCATORI = re.compile(r"\b(?:maybe|probably|possibly|perhaps|likely|presumably|apparently|i think|i believe|it seems|"
                       r"seems|appears|might|may|could)\b")
CONNETTIVO_O = re.compile(r"\bor\b|\beither\b|\bwhether\b")
# Correzione sui soli passaggi della prova: fonte generica, senza assegnazione dei due valori.
ATTRIBUZIONE_GENERICA = re.compile(
    r"\baccording to (?:the )?(?:different|separate|two|both) "
    r"(?:registers|archives|sources|records|chronicles)\b")

NUMERI_PAROLE = {w: i for i, w in enumerate(
    "zero one two three four five six seven eight nine ten eleven twelve thirteen fourteen fifteen sixteen seventeen "
    "eighteen nineteen twenty".split())}
NUMERI_PAROLE.update({"thirty": 30, "forty": 40, "fifty": 50, "sixty": 60, "seventy": 70, "eighty": 80, "ninety": 90})
TIPO_CONTORNO = ["anno", "anno", "numero", "numero", "luogo", "luogo", "persona", "persona", None, None]
PAROLE_FUNZIONE = set("the a an of in on at to for from by with and or is was were are be been it its this that these "
                      "those according answer passage text question saint doctor admiral lady city town company ship "
                      "festival archive register chronicle records river district fleet valley province year number "
                      "unknown yes no i".split())


def normalizza(s):
    """Normalizzazione SQuAD: minuscole, senza punteggiatura, senza articoli, spazi singoli."""
    s = s.lower()
    s = "".join(ch for ch in s if ch not in set(string.punctuation))
    s = re.sub(r"\b(a|an|the)\b", " ", s)
    return " ".join(s.split())


def pulisci(risposta):
    """Toglie markdown, virgolette e spazi ai bordi; tiene il testo della risposta così com'è."""
    return re.sub(r"[*_`#]+", "", risposta or "").strip().strip("\"'“”").strip()


def contiene_parola(testo_min, parola_min):
    return re.search(r"(?<![a-z0-9])" + re.escape(parola_min) + r"(?![a-z0-9])", testo_min) is not None


# --------------------------------------------------------------------------- valori nel testo
def valore_canonico(v, tipo):
    if tipo == "anno":
        return str(v)
    if tipo == "numero":
        return NUMERI_PAROLE.get(str(v).lower(), int(v) if str(v).isdigit() else str(v).lower())
    return str(v).lower()


def valori_nella_risposta(testo, tipo, nomi_noti):
    """Valori del tipo trovati nella risposta, in forma canonica. Per luoghi e persone: i nomi noti del passaggio trovati
    (un nome di più parole vale se compare una sua parola di almeno 4 lettere), più gli eventuali nomi sconosciuti."""
    t = testo.lower()
    if tipo == "anno":
        return {m for m in re.findall(r"(?<![0-9])(1[0-9]{3}|20[0-9]{2})(?![0-9])", t)}, set()
    if tipo == "numero":
        trovati = {int(m) for m in re.findall(r"(?<![0-9.,])([0-9]{1,6})(?![0-9])", t.replace(",", ""))}
        trovati |= {NUMERI_PAROLE[w] for w in re.findall(r"[a-z]+", t) if w in NUMERI_PAROLE}
        return trovati, set()
    trovati = set()
    for nome in nomi_noti:
        parti = [p for p in nome.lower().split() if len(p) >= 4] or [nome.lower()]
        if contiene_parola(t, nome.lower()) or any(contiene_parola(t, p) for p in parti):
            trovati.add(nome.lower())
    return trovati, set()


def nomi_sconosciuti(testo, esclusi):
    """Parole con la maiuscola nella risposta che non sono nomi del passaggio né parole di servizio."""
    out = set()
    for w in re.findall(r"\b[A-Z][a-zA-Z\-]+\b", testo):
        wl = w.lower()
        if wl in PAROLE_FUNZIONE or any(contiene_parola(e, wl) for e in esclusi):
            continue
        out.add(wl)
    return out


# --------------------------------------------------------------------------- insieme sintetico
def info_sintetico(passaggio, famiglia, lato):
    """Dalla rappresentazione formale di una versione: valori del tipo nel passaggio (bersaglio e contorno), nomi per
    l'attribuzione (fonti in CONF, gruppi in LOG) e tutti i nomi propri del passaggio."""
    tipo = passaggio["tipo_attributo"]
    rep = passaggio["famiglie"][famiglia][lato]["rep"]
    valori, tutti_nomi = set(), set()
    for f in rep:
        for k in ("v", "v1", "v2"):
            if k in f:
                valori.add(f[k])
        if f.get("tipo") == "contorno" and f.get("val") is not None:
            if TIPO_CONTORNO[f["i"]] == tipo:
                valori.add(f["val"])
            if TIPO_CONTORNO[f["i"]] in ("luogo", "persona"):
                tutti_nomi.add(str(f["val"]))
        for k in ("F", "G", "G1", "G2", "R"):
            if k in f:
                tutti_nomi.add(str(f[k]))
    nomi = passaggio["nomi"]
    attribuzione = nomi["fonti"] if famiglia == "CONF" else (nomi["gruppi"] if famiglia == "LOG" else [])
    tutti_nomi |= {passaggio["entita"], nomi.get("regione", "")} | set(nomi["fonti"]) | set(nomi["gruppi"])
    tutti_nomi |= {str(v) for v in valori if tipo in ("luogo", "persona")}
    return dict(tipo=tipo, famiglia=famiglia, lato=lato, v1=passaggio["valore_B"], v2=passaggio["valore_alternativo"],
                valori_contesto=sorted(str(v) for v in valori), attribuzione=list(attribuzione),
                nomi_passaggio=sorted(n for n in tutti_nomi if n))


def classifica_sintetico(risposta, info):
    """Categoria di §4 per una risposta su una versione sintetica. Restituisce (categoria, dettagli)."""
    testo = pulisci(risposta)
    t = testo.lower()
    tipo, fam, lato = info["tipo"], info["famiglia"], info["lato"]
    v1, v2 = valore_canonico(info["v1"], tipo), valore_canonico(info["v2"], tipo)
    ctx = {valore_canonico(v, tipo) for v in info["valori_contesto"]}
    nomi_tipo = [str(v) for v in info["valori_contesto"]] if tipo in ("luogo", "persona") else []
    trovati, _ = valori_nella_risposta(testo, tipo, nomi_tipo)
    det = dict(valori=sorted(map(str, trovati)))
    if not testo:
        return "altro", dict(det, motivo="risposta vuota")
    if ASTENSIONE.search(t):
        return "astensione", det
    if lato == "C" and fam != "RIM" and v1 in trovati and v2 in trovati:
        attr = info["attribuzione"]
        if fam in ("CONF", "LOG") and attr and all(contiene_parola(t, a.lower()) for a in attr):
            return "attribuite", det
        if CONNETTIVO_O.search(t):
            return "alternative", det
        if fam in ("CONF", "LOG") and ATTRIBUZIONE_GENERICA.search(t):
            return "altro", dict(det, motivo="attribuzione generica: assegnazione dei valori non esplicita")
        return "entrambi affermati", det
    if len(trovati) == 1 and MARCATORI.search(t):
        return "qualificata", det
    if len(trovati) == 1:
        (v,) = trovati
        if lato == "B":
            if v == v1:
                return "corretta", det
            return ("errata" if v in ctx else "valore assente dal contesto"), det
        return ("un valore del contesto" if v in ctx else "valore assente dal contesto"), det
    if len(trovati) > 1:
        return "altro", dict(det, motivo="più valori, non le due alternative")
    # nessun valore riconosciuto: un anno o un numero non c'è; per luoghi e persone si guardano i nomi sconosciuti
    if tipo in ("luogo", "persona"):
        ignoti = nomi_sconosciuti(testo, [n.lower() for n in info["nomi_passaggio"]])
        if ignoti and len(testo.split()) <= 8:
            if MARCATORI.search(t):
                return "qualificata", dict(det, nomi_sconosciuti=sorted(ignoti))
            return ("valore assente dal contesto"), dict(det, nomi_sconosciuti=sorted(ignoti))
    return "altro", dict(det, motivo="nessun valore del tipo riconosciuto")


# --------------------------------------------------------------------------- insieme naturale (descrittivo)
def classifica_naturale(risposta, riga):
    """Categorie di §4 per l'insieme naturale (descrittivo). B: corretta se una risposta di SQuAD e la risposta del
    modello si contengono dopo la normalizzazione. DUE-nat C: alternative o entrambi affermati se compaiono la risposta e
    il distrattore. RIM-nat C: ogni risposta non astenuta è non supportata; si segna se coincide con la risposta tolta
    (possibile memoria)."""
    testo = pulisci(risposta)
    t, n = testo.lower(), normalizza(testo)
    det = {}
    if not n:
        return "altro", dict(motivo="risposta vuota")
    if ASTENSIONE.search(t):
        return "astensione", det
    refs = [normalizza(a) for a in riga["risposte_B"] if normalizza(a)]
    giusta = any(contiene_parola(n, r) or contiene_parola(r, n) for r in refs)
    nel_contesto = contiene_parola(normalizza(riga["contesto"]), n)
    if riga["famiglia"] == "DUE-nat" and riga["kind"] == "C":
        d = normalizza(riga["distrattore"])
        con_d = contiene_parola(n, d) or contiene_parola(d, n)
        if giusta and con_d and len(n.split()) > max(len(d.split()), 1):
            return ("alternative" if CONNETTIVO_O.search(t) else "entrambi affermati"), det
        if MARCATORI.search(t) and (giusta or con_d):
            return "qualificata", det
        if giusta or con_d or nel_contesto:
            return "un valore del contesto", dict(det, distrattore=con_d)
        return "valore assente dal contesto", det
    if riga["kind"] == "B":
        if MARCATORI.search(t):
            return "qualificata", det
        if giusta:
            return "corretta", det
        return ("errata" if nel_contesto else "valore assente dal contesto"), det
    # RIM-nat, C
    if MARCATORI.search(t):
        return "qualificata", dict(det, risposta_tolta=giusta)
    if nel_contesto and not giusta:
        return "un valore del contesto", det
    return "valore assente dal contesto", dict(det, risposta_tolta=giusta)
```

```python
%%writefile /content/testC_run/script/tc_analisi.py
# Test C: misure e ipotesi (protocollo v3, §6–§7). Solo libreria standard e numpy: gira nel Python di Colab.
import math
from collections import Counter, defaultdict

import numpy as np

import tc_classifica as K

FAM_PRINCIPALI = ("DUE", "CONF", "LOG")


def finito(x):
    return x is not None and isinstance(x, (int, float)) and math.isfinite(x)


# --------------------------------------------------------------------------- statistica di base
def auroc(pos, neg):
    """AUROC come probabilità che un positivo abbia punteggio più alto di un negativo (pareggi a metà)."""
    pos, neg = np.asarray(pos, float), np.asarray(neg, float)
    if len(pos) == 0 or len(neg) == 0:
        return None
    allv = np.concatenate([pos, neg])
    order = allv.argsort(kind="mergesort")
    ranks = np.empty(len(allv))
    sv = allv[order]
    i = 0
    while i < len(sv):                           # ranghi medi per i pareggi
        j = i
        while j + 1 < len(sv) and sv[j + 1] == sv[i]:
            j += 1
        ranks[order[i:j + 1]] = (i + j) / 2 + 1
        i = j + 1
    rp = ranks[:len(pos)].sum()
    return float((rp - len(pos) * (len(pos) + 1) / 2) / (len(pos) * len(neg)))


def boot_media(valori_per_gruppo, seed, n_boot, smoke):
    """Media sui gruppi (passaggi) e IC al 95% ricampionando i gruppi."""
    x = np.asarray([v for v in valori_per_gruppo], float)
    n = len(x)
    if n == 0:
        return dict(passaggi=0, stima=None, ci95=None)
    ci = None
    if not smoke and n > 1:
        rng = np.random.default_rng(seed)
        b = x[rng.integers(0, n, (n_boot, n))].mean(axis=1)
        ci = [float(q) for q in np.quantile(b, [.025, .975])]
    return dict(passaggi=n, stima=float(x.mean()), ci95=ci)


def esito(stat, nullo, minimo, smoke=False):
    if smoke:
        return "PROVA TECNICA"
    if not stat or stat.get("ci95") is None:
        return "NON VALUTABILE"
    lo, hi = stat["ci95"]
    if lo > nullo and stat["stima"] >= minimo:
        return "CONFERMATA"
    return "NON CONFERMATA" if hi <= nullo else "NON CONCLUSIVA"


# --------------------------------------------------------------------------- coppie
def coppie(righe):
    """{pair_key: dict(passaggio, famiglia, B=row_id, C=row_id)} per le righe date."""
    out = {}
    for r in righe:
        d = out.setdefault(r["pair_key"], dict(passaggio=r["passaggio_id"], famiglia=r["famiglia"]))
        d[r["kind"]] = r["row_id"]
    return {k: v for k, v in out.items() if "B" in v and "C" in v}


def vittorie(punteggi, cp):
    """Per ogni coppia: 1 se il punteggio è più alto su C, 0.5 se uguale, 0 se più basso; None se manca."""
    out = {}
    for k, d in cp.items():
        b, c = punteggi.get(d["B"]), punteggi.get(d["C"])
        out[k] = None if not (finito(b) and finito(c)) else (1.0 if c > b else 0.5 if c == b else 0.0)
    return out


def W(punteggi, cp, famiglie, seed, n_boot, smoke, minimo_coppie=None):
    v = vittorie(punteggi, cp)
    per_p = defaultdict(list)
    tot, valide, escluse = 0, 0, Counter()
    for k, d in cp.items():
        if d["famiglia"] not in famiglie:
            continue
        tot += 1
        if v[k] is None:
            escluse[d["famiglia"]] += 1
            continue
        valide += 1
        per_p[d["passaggio"]].append(v[k])
    stat = boot_media([np.mean(x) for x in per_p.values()], seed, n_boot, smoke)
    stat["copertura"] = dict(coppie=tot, valide=valide, escluse_per_famiglia=dict(escluse), minimo=minimo_coppie,
                             sufficiente=(minimo_coppie is None or valide >= minimo_coppie))
    return stat


def delta_W(punteggi, rif, cp, famiglie, seed, n_boot, smoke):
    """ΔW sulle coppie valide per entrambi, ricampionando lo stesso passaggio nei due bracci."""
    v, vr = vittorie(punteggi, cp), vittorie(rif, cp)
    per_p, per_r, per_m = defaultdict(list), defaultdict(list), defaultdict(list)
    n = 0
    for k, d in cp.items():
        if d["famiglia"] not in famiglie or v[k] is None or vr[k] is None:
            continue
        n += 1
        per_p[d["passaggio"]].append(v[k] - vr[k])
        per_r[d["passaggio"]].append(vr[k])
        per_m[d["passaggio"]].append(v[k])
    stat = boot_media([np.mean(x) for x in per_p.values()], seed, n_boot, smoke)
    w_rif = float(np.mean([np.mean(x) for x in per_r.values()])) if per_r else None
    stat.update(coppie=n, W_riferimento_stesse_coppie=w_rif,
                W_metodo_stesse_coppie=float(np.mean([np.mean(x) for x in per_m.values()])) if per_m else None,
                margine=(1 - w_rif) if w_rif is not None else None)
    return stat


def D_rim(punteggi, cp, seed, n_boot, smoke):
    """HC3: D = W(RIM) − W(DUE+CONF+LOG), per passaggio, sugli stessi passaggi."""
    v = vittorie(punteggi, cp)
    rim, princ = {}, defaultdict(list)
    for k, d in cp.items():
        if v[k] is None:
            continue
        if d["famiglia"] == "RIM":
            rim[d["passaggio"]] = v[k]
        elif d["famiglia"] in FAM_PRINCIPALI:
            princ[d["passaggio"]].append(v[k])
    comuni = sorted(set(rim) & set(princ))
    return boot_media([rim[p] - np.mean(princ[p]) for p in comuni], seed, n_boot, smoke)


# --------------------------------------------------------------------------- HC4
def auroc_per_passaggio(pos, neg, seed, n_boot, smoke, minimo_risposte, minimo_passaggi):
    """pos, neg: liste di (passaggio, punteggio). AUROC con IC ricampionando i passaggi (tutte le risposte di un
    passaggio entrano o escono insieme). NON VALUTABILE sotto le numerosità minime (§6.2)."""
    pos = [(p, s) for p, s in pos if finito(s)]
    neg = [(p, s) for p, s in neg if finito(s)]
    np_pos, np_neg = len({p for p, _ in pos}), len({p for p, _ in neg})
    out = dict(n_pos=len(pos), n_neg=len(neg), passaggi_pos=np_pos, passaggi_neg=np_neg,
               stima=auroc([s for _, s in pos], [s for _, s in neg]), ci95=None)
    valutabile = (len(pos) >= minimo_risposte and len(neg) >= minimo_risposte and np_pos >= minimo_passaggi
                  and np_neg >= minimo_passaggi)
    out["valutabile"] = valutabile
    if smoke or not valutabile or out["stima"] is None:
        return out
    passaggi = sorted({p for p, _ in pos} | {p for p, _ in neg})
    idx = {p: i for i, p in enumerate(passaggi)}
    gp, gn = defaultdict(list), defaultdict(list)
    for p, s in pos:
        gp[idx[p]].append(s)
    for p, s in neg:
        gn[idx[p]].append(s)
    rng = np.random.default_rng(seed)
    stime = []
    for _ in range(n_boot):
        scelta = rng.integers(0, len(passaggi), len(passaggi))
        a = [s for i in scelta for s in gp.get(i, ())]
        b = [s for i in scelta for s in gn.get(i, ())]
        if a and b:
            stime.append(auroc(a, b))
    if stime:
        out["ci95"] = [float(q) for q in np.quantile(stime, [.025, .975])]
    return out


def esito_hc4(stat, smoke):
    if smoke:
        return "PROVA TECNICA"
    if not stat.get("valutabile"):
        return "NON VALUTABILE"
    return esito(dict(stima=stat["stima"], ci95=stat["ci95"]), 0.5, 0.70)


def insiemi_hc4(righe, categorie, punteggi):
    """Le classi di §6.2. righe: le righe sintetiche; categorie: {row_id: categoria}; punteggi: {row_id: valore}."""
    def lista(filtro):
        return [(r["passaggio_id"], punteggi.get(r["row_id"])) for r in righe if filtro(r, categorie.get(r["row_id"]))]
    in_c = lambda r: r["kind"] == "C" and r["famiglia"] in FAM_PRINCIPALI
    in_b = lambda r: r["kind"] == "B"
    return dict(
        entro_C=(lista(lambda r, c: in_c(r) and c in K.NON_SUPPORTATE_C),
                 lista(lambda r, c: in_c(r) and c in K.SUPPORTATE_C)),
        entro_B=(lista(lambda r, c: in_b(r) and c in K.NON_SUPPORTATE_B),
                 lista(lambda r, c: in_b(r) and c in K.SUPPORTATE_B)),
        # descrittivi (§6.2)
        entro_C_con_astensioni=(lista(lambda r, c: in_c(r) and c in K.NON_SUPPORTATE_C),
                                lista(lambda r, c: in_c(r) and (c in K.SUPPORTATE_C or c == "astensione"))),
        C_non_supportate_contro_B_corrette=(lista(lambda r, c: in_c(r) and c in K.NON_SUPPORTATE_C),
                                            lista(lambda r, c: in_b(r) and r["famiglia"] in FAM_PRINCIPALI
                                                  and c in K.SUPPORTATE_B)),
        C_contro_B_con_astensioni=(lista(lambda r, c: in_c(r) and c in K.NON_SUPPORTATE_C),
                                   lista(lambda r, c: (in_b(r) and r["famiglia"] in FAM_PRINCIPALI
                                                       and c in K.SUPPORTATE_B) or (in_c(r) and c == "astensione"))),
    )


def conteggi_categorie(righe, categorie):
    out = defaultdict(Counter)
    for r in righe:
        out[f"{r['insieme']}|{r['famiglia']}|{r['kind']}"][categorie.get(r["row_id"], "mancante")] += 1
    return {k: dict(v) for k, v in sorted(out.items())}


def accuratezza_B(righe, categorie):
    """HC0-set (descrittivo): quota di risposte corrette su B, per famiglia e in tutto (astensioni comprese nel totale)."""
    per = defaultdict(lambda: [0, 0])
    for r in righe:
        if r["kind"] != "B":
            continue
        for k in (r["famiglia"], "tutte"):
            per[k][1] += 1
            per[k][0] += categorie.get(r["row_id"]) == "corretta"
    return {k: dict(corrette=a, totale=n, accuratezza=(a / n if n else None)) for k, (a, n) in sorted(per.items())}
```

```python
%%writefile /content/testC_run/script/tc_support.py
"""Esportazione e provenienza del Test C: nessuna inferenza o formula statistica."""
import csv
import hashlib
import json
import random
from collections import defaultdict
from pathlib import Path


def response_series(load, classify, models):
    series = []
    for model in models:
        data = load(f"M1_{model}")
        if data is None:
            continue
        for source, key, field in (("M1-riferimento", "riferimento", "generated"),
                                   ("M1-LM-Polygraph", "polygraph", "greedy_polygraph")):
            answers = {r["row_id"]: r[field] for r in data[key]}
            categories, details = classify(answers)
            series.append(dict(modello=model, origine=source, risposte=answers, categorie=categories, dettagli=details))
    data = load("M4_llama")
    if data is not None:
        answers = {r["row_id"]: r["generated"] for r in data["generazione"]}
        categories, details = classify(answers)
        series.append(dict(modello="llama", origine="M4-generazione", risposte=answers,
                           categorie=categories, dettagli=details))
    return series


def write_responses(res, rows, root, write_json):
    root = Path(root)
    by = {r["row_id"]: r for r in rows}
    records = []
    for series in res.get("serie_risposte", []):
        for rid, text in series["risposte"].items():
            r = by[rid]
            records.append(dict(modello=series["modello"], origine=series["origine"], row_id=rid,
                                insieme=r["insieme"], famiglia=r["famiglia"], lato=r["kind"],
                                tipo=r["tipo_attributo"], contesto=r["contesto"], domanda=r["domanda"], risposta=text,
                                categoria=series["categorie"].get(rid),
                                dettagli=json.dumps(series["dettagli"].get(rid, {}), ensure_ascii=False)))
    columns = ["modello", "origine", "row_id", "insieme", "famiglia", "lato", "tipo", "contesto", "domanda",
               "risposta", "categoria", "dettagli"]
    with open(root / "risposte_C.csv", "w", newline="") as fh:
        writer = csv.DictWriter(fh, fieldnames=columns)
        writer.writeheader()
        writer.writerows(records)
    # Il campione copre anche M1 e M4. La categoria automatica non viene mostrata nella scheda da annotare.
    rng = random.Random(20261115)
    strata = defaultdict(list)
    for r in records:
        if r["insieme"].startswith("sintetico"):
            strata[(r["categoria"] or "", r["origine"], r["modello"])].append(r)
    chosen = []
    keys = sorted(strata)
    while len(chosen) < 100 and any(strata[k] for k in keys):
        for k in keys:
            if strata[k] and len(chosen) < 100:
                chosen.append(strata[k].pop(rng.randrange(len(strata[k]))))
    rng.shuffle(chosen)
    blind = ["n", "modello", "origine", "row_id", "famiglia", "lato", "contesto", "domanda", "risposta",
             "categoria_Simona"]
    with open(root / "campione_classificazione.csv", "w", newline="") as fh:
        writer = csv.DictWriter(fh, fieldnames=blind)
        writer.writeheader()
        for i, r in enumerate(chosen, 1):
            writer.writerow(dict(n=i, categoria_Simona="", **{k: r[k] for k in blind if k not in
                                                                           ("n", "categoria_Simona")}))
    write_json(root / "campione_classificazione_chiave.json",
               [dict(n=i, row_id=r["row_id"], modello=r["modello"], origine=r["origine"], categoria=r["categoria"])
                for i, r in enumerate(chosen, 1)])
    by_source = {(s["modello"], s["origine"]): s for s in res.get("serie_risposte", [])}
    comparisons = []
    for model in sorted({m for m, _ in by_source}):
        ref = by_source.get((model, "M1-riferimento"))
        poly = by_source.get((model, "M1-LM-Polygraph"))
        if ref is None or poly is None:
            continue
        for rid in sorted(ref["risposte"].keys() & poly["risposte"].keys()):
            a, b = ref["risposte"][rid], poly["risposte"][rid]
            comparisons.append(dict(modello=model, row_id=rid, riferimento=a, LM_Polygraph=b,
                                    testo_uguale=a.strip() == b.strip(),
                                    categoria_uguale=ref["categorie"][rid] == poly["categorie"][rid]))
    with open(root / "confronto_M1_riferimento.csv", "w", newline="") as fh:
        writer = csv.DictWriter(fh, fieldnames=["modello", "row_id", "riferimento", "LM_Polygraph", "testo_uguale",
                                               "categoria_uguale"])
        writer.writeheader()
        writer.writerows(comparisons)


def job_provenance(state, load_json, session_gpu):
    result = {}
    for job, info in state.items():
        if job.startswith("installazione"):
            continue
        raw_gpu = None
        p = Path(info["cartella"]) / info["file"] if info.get("cartella") and info.get("file") else None
        error = None
        if p is not None and p.is_file():
            try:
                raw_gpu = load_json(p).get("gpu")
            except (OSError, ValueError, TypeError) as exc:
                error = str(exc)
        result[job] = dict(gpu_inferenza=raw_gpu, gpu_sessione_corrente=session_gpu,
                           gpu_stato_precedente=info.get("gpu"),
                           fonte_gpu_inferenza="uscita grezza" if raw_gpu else "non disponibile",
                           errore_lettura=error)
    return result


def bundle_inputs(z, rows, passages, data_dir, state, hc0_sources, config, scripts):
    """Aggiunge input/configurazioni, senza ambienti, credenziali o checkpoint delle sonde."""
    index, written = [], set()

    def put(name, blob):
        if name in written:
            return
        z.writestr(name, blob)
        written.add(name)
        index.append(dict(file=name, bytes=len(blob), sha256=hashlib.sha256(blob).hexdigest()))

    def put_json(name, value):
        put(name, json.dumps(value, sort_keys=True, ensure_ascii=False, allow_nan=False).encode("utf-8"))

    for path in sorted(Path(data_dir).glob("*.json")):
        put("dati/" + path.name, path.read_bytes())
    put_json("dati/righe_C_selezionate.json", rows)
    ids = {r["passaggio_id"] for r in rows if r["insieme"].startswith("sintetico")}
    put_json("dati/passaggi_selezionati.json", {k: v for k, v in passages.items() if k in ids})
    put_json("configurazione_notebook.json", config)
    for job, info in state.items():
        folder = info.get("cartella")
        if not folder:
            continue
        path = Path(folder) / "config.json"
        if path.is_file():
            put("config_lavori/" + job + ".json", path.read_bytes())
        else:
            index.append(dict(file="config_lavori/" + job + ".json", stato="mancante"))
    for key, info in hc0_sources.items():
        path = Path(info["file"])
        if info.get("presente") and path.is_file():
            put("riferimenti_hc0/esiti_" + key + ".json", path.read_bytes())
    for path in sorted(Path(scripts).glob("*.py")):
        put("script/" + path.name, path.read_bytes())
    put_json("manifest_input.json", index)

```

```python
# 10. Analisi (protocollo v3, §4–§7) ed esportazione. Gira nel Python di Colab e legge solo i file su Drive.
import csv
import importlib
import random
import traceback
import zipfile
from collections import Counter, defaultdict

import tc_classifica as K
import tc_analisi as A
import tc_support as S
importlib.reload(K)
importlib.reload(A)
importlib.reload(S)

STIMATORI_M1 = ["MSP", "Perplexity", "MeanTokenEntropy", "SemanticEntropy", "PTrue"]
NOME_SONDA = {"llama3_1_8b_linear": "M4-lin", "llama3_1_8b_lora_lambda_kl_0_05": "M4-LoRA"}

def carica(job):
    f = job_file(job)
    return read_json(f) if f is not None else None

def esiti_hc0():
    # HC0 ereditato dal Test M: {(metodo, modello): dict(superato, valore, misura, fonte)}
    out, fonti = {}, {}
    for nome, path in (("A", ESITI_A), ("B", ESITI_B)):
        p = Path(path)
        if not p.is_file():
            fonti[nome] = dict(file=path, presente=False)
            continue
        e = read_json(p)
        fonti[nome] = dict(file=path, presente=True, sha256=sha_file(p), versione=e.get("versione"),
                           modalita=e.get("modalita"))
        for r in e.get("risultati", []):
            modello = "giudice" if r["metodo"] == "M3" else r["modello"]
            h = r.get("HM0", {})
            out[(r["metodo"], modello)] = dict(superato=bool(h.get("superato")), valore=h.get("valore"),
                                               misura=h.get("misura"), esito=h.get("esito"), fonte=f"esiti_{nome}")
    return out, fonti

def righe_sintetiche():
    return [r for r in RIGHE if r["insieme"].startswith("sintetico")]

def classifica_risposte(risposte):
    # risposte: {row_id: testo}; restituisce {row_id: categoria} e {row_id: dettagli}
    cat, det = {}, {}
    for r in RIGHE:
        if r["row_id"] not in risposte:
            continue
        if r["insieme"].startswith("sintetico"):
            info = K.info_sintetico(PASSAGGI[r["passaggio_id"]], r["famiglia"], r["kind"])
            c, d = K.classifica_sintetico(risposte[r["row_id"]], info)
        else:
            c, d = K.classifica_naturale(risposte[r["row_id"]], r)
        cat[r["row_id"]], det[r["row_id"]] = c, d
    return cat, det

def misure_metodo(nome, modello, punteggi, rif, categorie, hc0, sint, nat, cp_s, cp_n):
    # W, ΔW, HC1–HC4 per un metodo e un modello generatore (§6–§7)
    n_main = sum(d["famiglia"] in A.FAM_PRINCIPALI for d in cp_s.values())
    n_rim = sum(d["famiglia"] == "RIM" for d in cp_s.values())
    m_main = None if SMOKE else int(np.ceil(QUOTA_COPERTURA * n_main))
    m_rim = None if SMOKE else int(np.ceil(QUOTA_COPERTURA * n_rim))
    ok0 = SMOKE or bool(hc0 and hc0["superato"])
    out = dict(metodo=nome, modello=modello, HC0=hc0)
    w = A.W(punteggi, cp_s, A.FAM_PRINCIPALI, SEED, N_BOOT, SMOKE, m_main)
    out["W"] = w
    out["HC1"] = (A.esito(w, 0.5, 0.60, SMOKE) if ok0 and w["copertura"]["sufficiente"] else "NON VALUTABILE")
    if rif is not None:
        dw = A.delta_W(punteggi, rif, cp_s, A.FAM_PRINCIPALI, SEED, N_BOOT, SMOKE)
        out["delta_W"] = dw
        out["HC2"] = (A.esito(dw, 0.0, 0.05, SMOKE) if ok0 and (SMOKE or dw["coppie"] >= m_main) else "NON VALUTABILE")
    w_rim = A.W(punteggi, cp_s, ("RIM",), SEED, N_BOOT, SMOKE, m_rim)
    out["W_RIM"] = w_rim
    out["HC_dom"] = A.esito(w_rim, 0.5, 0.60, SMOKE) if w_rim["copertura"]["sufficiente"] else "NON VALUTABILE"
    d = A.D_rim(punteggi, cp_s, SEED, N_BOOT, SMOKE)
    out["D_RIM"] = d
    out["HC3"] = A.esito(d, 0.0, 0.10, SMOKE) if ok0 and w["copertura"]["sufficiente"] else "NON VALUTABILE"
    out["famiglie"] = {}
    for f in A.FAM_PRINCIPALI:
        m_f = None if SMOKE else int(np.ceil(QUOTA_COPERTURA * sum(x["famiglia"] == f for x in cp_s.values())))
        wf = A.W(punteggi, cp_s, (f,), SEED, N_BOOT, SMOKE, m_f)
        ff = dict(W=wf, HC1_nominale=A.esito(wf, 0.5, 0.60, SMOKE) if ok0 and wf["copertura"]["sufficiente"]
                  else "NON VALUTABILE")
        if rif is not None:
            ff["delta_W"] = A.delta_W(punteggi, rif, cp_s, (f,), SEED, N_BOOT, SMOKE)
        out["famiglie"][f] = ff
    if categorie is not None:
        ins = A.insiemi_hc4(sint, categorie, punteggi)
        hc4 = {}
        for k, (pos, neg) in ins.items():
            st = A.auroc_per_passaggio(pos, neg, SEED, N_BOOT, SMOKE, MIN_RISPOSTE_HC4, MIN_PASSAGGI_HC4)
            if k in ("entro_C", "entro_B"):
                st["esito"] = A.esito_hc4(st, SMOKE) if ok0 else "NON VALUTABILE"
            hc4[k] = st
        out["HC4"] = hc4
    if cp_n:
        out["naturale"] = {f: A.W(punteggi, cp_n, (f,), SEED, N_BOOT, True) for f in ("RIM-nat", "DUE-nat")}
        if rif is not None:
            out["naturale_delta_W"] = {f: A.delta_W(punteggi, rif, cp_n, (f,), SEED, N_BOOT, True)
                                       for f in ("RIM-nat", "DUE-nat")}
    return out

def analizza_tutto():
    errori, risultati, comportamento, descr = [], [], {}, {}
    hc0, fonti_hc0 = esiti_hc0()
    sint = righe_sintetiche()
    nat = [r for r in RIGHE if r["insieme"] == "naturale"]
    cp_s, cp_n = A.coppie(sint), A.coppie(nat)
    risposte_canon, categorie_canon, dettagli_canon = {}, {}, {}
    rif_m1, rif_sirin = {}, {}
    m3 = None
    try:
        m3 = carica("M3_giudice")
    except Exception:                            # noqa: BLE001
        errori.append(f"M3: {traceback.format_exc()[-1500:]}")
    for key in MODEL_CONFIG:
        try:
            d1 = carica(f"M1_{key}")
            rs = carica(f"REF_env_sirin_{key}")
            if rs is not None:
                rif_sirin[key] = {r["row_id"]: r["p_UNKNOWN"] for r in rs["riferimento"]}
            if d1 is None:
                continue
            ref = {r["row_id"]: r for r in d1["riferimento"]}
            rif_m1[key] = {k: r["p_UNKNOWN"] for k, r in ref.items()}
            risposte_canon[key] = {k: r["generated"] for k, r in ref.items()}
            categorie_canon[key], dettagli_canon[key] = classifica_risposte(risposte_canon[key])
            gen = list(ref.values())
            comportamento[key] = dict(
                categorie=A.conteggi_categorie(RIGHE, categorie_canon[key]),
                accuratezza_B_sintetico=A.accuratezza_B(sint, categorie_canon[key]),
                accuratezza_B_naturale=A.accuratezza_B(nat, categorie_canon[key]),
                risposte_finite_entro_il_limite=sum(g["fine_naturale"] for g in gen) / len(gen),
                token_generati_medi=float(np.mean([g["n_token_generati"] for g in gen])),
                allineamento_U2_env_m1=d1["allineamento"], greedy_LM_Polygraph_uguale=d1["greedy_uguale_al_riferimento"],
                quota_altro_sintetico=(sum(categorie_canon[key].get(r["row_id"]) == "altro" for r in sint) / len(sint))
                if sint else None)
            acc = comportamento[key]["accuratezza_B_sintetico"].get("tutte", {}).get("accuratezza")
            comportamento[key]["HC0_set_avviso"] = (acc is not None and acc < SOGLIA_HC0_SET)
            if rs is not None:
                comune = [k for k in rif_m1[key] if k in rif_sirin[key]]
                comportamento[key]["p_UNKNOWN_env_sirin_meno_env_m1_max"] = max(
                    abs(rif_sirin[key][k] - rif_m1[key][k]) for k in comune) if comune else None
                comportamento[key]["allineamento_U2_env_sirin"] = rs["allineamento"]
            # riferimento p(UNKNOWN) di env_m1: descrittivo (è il termine di confronto, non un metodo)
            risultati.append(misure_metodo("riferimento p(UNKNOWN)", key, rif_m1[key], None, categorie_canon[key],
                                           dict(superato=True, esito="riferimento"), sint, nat, cp_s, cp_n))
            # M1: le etichette di HC4 vengono dalla risposta che LM-Polygraph ha valutato
            poly = {r["row_id"]: r for r in d1["polygraph"]}
            cat_poly, _ = classifica_risposte({k: r["greedy_polygraph"] for k, r in poly.items()})
            comuni = sorted(ref.keys() & poly.keys())
            comportamento[key]["confronto_M1_riferimento"] = dict(
                prompt_comuni=len(comuni),
                testo_uguale=sum(poly[k]["greedy_polygraph"].strip() == ref[k]["generated"].strip() for k in comuni),
                categoria_uguale=sum(cat_poly[k] == categorie_canon[key][k] for k in comuni),
                id_testo_diverso=[k for k in comuni if poly[k]["greedy_polygraph"].strip() != ref[k]["generated"].strip()])
            for est in STIMATORI_M1:
                sc = {k: r[f"u_{est}"] for k, r in poly.items()}
                risultati.append(misure_metodo(f"M1-{est}", key, sc, rif_m1[key], cat_poly,
                                               hc0.get((f"M1-{est}", key)), sint, nat, cp_s, cp_n))
        except Exception:                        # noqa: BLE001
            errori.append(f"M1/riferimento {key}: {traceback.format_exc()[-1500:]}")
    for key in MODEL_CONFIG:
        try:
            d2 = carica(f"M2_{key}")
            if d2 is None:
                continue
            for var, metodo in (("originale", "M2"), ("senza_l2", "M2-senzaℓ2")):
                ctrl = d2["controllo_ricaricamento"].get(var, {})
                if not ctrl.get("superato") or var not in d2["righe"]:
                    descr[f"{metodo}|{key}"] = dict(motivo="controllo del ricaricamento non superato", controllo=ctrl)
                    continue
                sc = {r["row_id"]: r["p_non_risp"] for r in d2["righe"][var]}
                res = misure_metodo(metodo, key, sc, rif_sirin.get(key), categorie_canon.get(key),
                                    hc0.get((metodo, key)), sint, nat, cp_s, cp_n)
                res["controllo_ricaricamento"] = ctrl
                risultati.append(res)
        except Exception:                        # noqa: BLE001
            errori.append(f"M2 {key}: {traceback.format_exc()[-1500:]}")
    if m3 is not None:
        try:
            sc = {r["row_id"]: r["p_non_risp"] for r in m3["righe"]}
            # un punteggio per prompt: HC1, HC3 e HC-dom valgono una volta; HC2 e HC4 per ciascun modello generatore
            for key in MODEL_CONFIG:
                if key not in categorie_canon and key not in rif_sirin:
                    continue
                risultati.append(misure_metodo("M3", key, sc, rif_sirin.get(key), categorie_canon.get(key),
                                               hc0.get(("M3", "giudice")), sint, nat, cp_s, cp_n))
        except Exception:                        # noqa: BLE001
            errori.append(f"M3: {traceback.format_exc()[-1500:]}")
    try:
        d4 = carica("M4_llama")
        if d4 is not None:
            gen = {g["row_id"]: g for g in d4["generazione"]}
            rif_obeso = {k: g["p_UNKNOWN"] for k, g in gen.items()}
            cat4, _ = classifica_risposte({k: g["generated"] for k, g in gen.items()})
            if "llama" in risposte_canon:
                uguali = sum(gen[k]["generated"].strip() == v.strip() for k, v in risposte_canon["llama"].items() if k in gen)
                descr["M4_risposte_uguali_a_env_m1"] = dict(uguali=uguali, n=len(gen))
            for pid, rows in d4["sonde"].items():
                sc = {r["row_id"]: r["sonda_max"] for r in rows}
                risultati.append(misure_metodo(NOME_SONDA[pid], "llama", sc, rif_obeso, cat4,
                                               hc0.get((NOME_SONDA[pid], "llama")), sint, nat, cp_s, cp_n))
            descr["M4_allineamento_U2"] = d4["allineamento"]
    except Exception:                            # noqa: BLE001
        errori.append(f"M4: {traceback.format_exc()[-1500:]}")
    serie = []
    try:
        serie = S.response_series(carica, classifica_risposte, MODEL_CONFIG)
    except Exception:
        errori.append(f"serie risposte: {traceback.format_exc()[-1500:]}")
    return dict(risultati=risultati, comportamento=comportamento, descrittivi=descr, errori=errori, hc0_fonti=fonti_hc0,
                risposte=risposte_canon, categorie=categorie_canon, dettagli=dettagli_canon, serie_risposte=serie)

def lettura(risultati):
    if SMOKE:
        return ["Prova tecnica: nessun conteggio di conferme o esito scientifico."]
    # regola dei 3 su 4 modelli valutabili (§7); M3: HC1 e HC3 una volta, HC2 e HC4 per generatore; M4 solo Llama
    righe = []
    per = defaultdict(list)
    for r in risultati:
        if r["metodo"] != "riferimento p(UNKNOWN)":
            per[r["metodo"]].append(r)
    for metodo, rs in per.items():
        val = [r for r in rs if r["HC0"] and r["HC0"].get("superato")]
        if metodo == "M3":
            r0 = rs[0]
            righe.append(f"M3 (giudice unico): HC1 {r0['HC1']}, HC3 {r0['HC3']}, HC-dom {r0['HC_dom']}; "
                         f"HC2 confermata con il p(UNKNOWN) di {sum(r.get('HC2') == 'CONFERMATA' for r in rs)} modelli su "
                         f"{len(rs)} (quattro confronti dello stesso giudice); HC4 entro C confermata per "
                         f"{sum(r['HC4']['entro_C'].get('esito') == 'CONFERMATA' for r in rs if 'HC4' in r)} generatori, "
                         f"entro B per {sum(r['HC4']['entro_B'].get('esito') == 'CONFERMATA' for r in rs if 'HC4' in r)}.")
            continue
        conta = {h: sum(r.get(h) == "CONFERMATA" for r in val) for h in ("HC1", "HC2", "HC3")}
        c4 = {k: sum(r.get("HC4", {}).get(k, {}).get("esito") == "CONFERMATA" for r in val) for k in ("entro_C", "entro_B")}
        v4 = {k: sum(r.get("HC4", {}).get(k, {}).get("esito") not in (None, "NON VALUTABILE") for r in val)
              for k in ("entro_C", "entro_B")}
        soglia = lambda n: n >= 3 if len(val) >= 3 else False
        frase = (f"{metodo}: valutabile in {len(val)} modelli; HC1 confermata in {conta['HC1']}, HC2 in {conta['HC2']}, "
                 f"HC3 in {conta['HC3']}; HC4 entro C in {c4['entro_C']} su {v4['entro_C']} valutabili, entro B in "
                 f"{c4['entro_B']} su {v4['entro_B']}.")
        if soglia(conta["HC1"]):
            frase += f" {metodo} riconosce l'indeterminatezza nel contesto (3 modelli su 4)."
        if v4["entro_C"] >= 3 and c4["entro_C"] >= 3:
            frase += f" {metodo} discrimina il supporto fra le risposte non astenute (entro C)."
        for r in val:
            if r["HC1"] == "NON CONFERMATA":
                frase += f" {r['modello']}: funziona sul suo dominio ma non riconosce l'indeterminatezza nel contesto."
            elif r["HC1"] == "NON CONCLUSIVA":
                frase += f" {r['modello']}: non ha superato il criterio HC1 in questo test."
        righe.append(frase)
    return righe

def stat_txt(s):
    if not s or s.get("stima") is None:
        return "—"
    ci = s.get("ci95")
    return f"{s['stima']:.3f}" + (f" [{ci[0]:.3f}, {ci[1]:.3f}]" if ci else "")

COLONNE = ["metodo", "modello", "HC0", "W", "W_ci_basso", "W_ci_alto", "coppie_valide", "HC1", "delta_W", "dW_ci_basso",
           "dW_ci_alto", "margine", "HC2", "W_RIM", "HC_dom", "D_RIM", "HC3", "AUROC_entro_C", "n_C_pos", "n_C_neg",
           "HC4_entro_C", "AUROC_entro_B", "n_B_pos", "n_B_neg", "HC4_entro_B"]

def riga_tabella(r):
    g = lambda s, i: (s.get("ci95") or [None, None])[i] if s else None
    h4 = r.get("HC4", {})
    return dict(metodo=r["metodo"], modello=r["modello"], HC0=(r["HC0"] or {}).get("valore"), W=r["W"]["stima"],
                W_ci_basso=g(r["W"], 0), W_ci_alto=g(r["W"], 1), coppie_valide=r["W"]["copertura"]["valide"],
                HC1=r["HC1"], delta_W=r.get("delta_W", {}).get("stima"), dW_ci_basso=g(r.get("delta_W"), 0),
                dW_ci_alto=g(r.get("delta_W"), 1), margine=r.get("delta_W", {}).get("margine"), HC2=r.get("HC2"),
                W_RIM=r["W_RIM"]["stima"], HC_dom=r["HC_dom"], D_RIM=r["D_RIM"]["stima"], HC3=r["HC3"],
                AUROC_entro_C=h4.get("entro_C", {}).get("stima"), n_C_pos=h4.get("entro_C", {}).get("n_pos"),
                n_C_neg=h4.get("entro_C", {}).get("n_neg"), HC4_entro_C=h4.get("entro_C", {}).get("esito"),
                AUROC_entro_B=h4.get("entro_B", {}).get("stima"), n_B_pos=h4.get("entro_B", {}).get("n_pos"),
                n_B_neg=h4.get("entro_B", {}).get("n_neg"), HC4_entro_B=h4.get("entro_B", {}).get("esito"))

def scrivi_risposte(res):
    S.write_responses(res, RIGHE, RADICE, write_json)

def export():
    st = stato()
    try:
        res = analizza_tutto()
    except Exception:
        res = dict(risultati=[], comportamento={}, descrittivi={}, hc0_fonti={}, serie_risposte=[],
                   errori=[f"analisi interrotta: {traceback.format_exc()[-3000:]}"])
    provenance = S.job_provenance(st, read_json, GPU)
    out = dict(versione=VERSIONE, selezione=VERSIONE_SELEZIONE, regole_naturale=REGOLE_NATURALE, sha_zip_selezione=SHA_ARCHIVIO_SELEZIONE, regole_classificazione=K.VERSIONE_REGOLE, modalita=MODALITA, dati=DATA_SHA,
               sha_righe_C=SHA_RIGHE_C, stato=st, gpu_sessione=GPU, provenienza_lavori=provenance, m1_generazione=M1_GENERAZIONE, hc0_fonti=res["hc0_fonti"],
               comportamento=res["comportamento"], descrittivi=res["descrittivi"], risultati=res["risultati"],
               lettura=lettura(res["risultati"]), errori_analisi=res["errori"])
    pd.DataFrame([riga_tabella(r) for r in res["risultati"]], columns=COLONNE).to_csv(RADICE / "tabella_C.csv", index=False)
    try:
        scrivi_risposte(res)
    except Exception:                            # noqa: BLE001
        res["errori"].append(f"risposte: {traceback.format_exc()[-1500:]}")
    write_json(RADICE / "esiti_C.json", out)
    linee = [f"Test C · {VERSIONE} · {MODALITA} · selezione {VERSIONE_SELEZIONE} · regole {K.VERSIONE_REGOLE} · righe {len(RIGHE)} ({DATA_SHA[:12]})", "",
             "Stato dei lavori:"]
    for job, info in sorted(st.items()):
        motivo = info.get("motivo") or (f"codice {info['codice']}" if "codice" in info else "") or \
            (info.get("errore") or "")[-200:]
        usato = "" if info.get("esito") != "ESEGUITO" or job.startswith("installazione") else \
            (" (usato nell'analisi)" if job_file(job) is not None else " (NON usato: impronta diversa da quella attesa)")
        linee.append(f"  {job}: {info.get('esito')}{usato}" + (f" — {motivo}" if motivo and info.get("esito") != "ESEGUITO" else ""))
    linee += ["", f"Dataset: {VERSIONE_SELEZIONE}/{REGOLE_NATURALE}; SHA ZIP originale {SHA_ARCHIVIO_SELEZIONE}", "Deficit naturale DUE-nat: 199/200, dichiarato (RIM-nat 200/200), insieme descrittivo.", f"GPU della sessione corrente: {GPU or 'nessuna'}"]
    for job, prov in sorted(provenance.items()):
        linee.append(f"  {job}: GPU inferenza {prov['gpu_inferenza'] or 'non disponibile'} "
                     f"(fonte: {prov['fonte_gpu_inferenza']})")
    linee += ["", "M1: stop naturale, nessun minimo forzato, limite TOKEN_RISPOSTA della configurazione. "
              "Deviazione dal minimo di 2 token di LM-Polygraph 0.7.0, anche nelle generazioni ausiliarie.",
              "HC0 di M1: ereditato dal Test M con configurazione precedente; "
              "verificarne l’applicabilita prima del completo."]
    for serie in res.get("serie_risposte", []):
        counts = Counter(serie["categorie"].values())
        linee.append(f"Categorie {serie['modello']} / {serie['origine']}: {dict(sorted(counts.items()))}")
    linee.append("")
    for key, c in res["comportamento"].items():
        cmp = c.get("confronto_M1_riferimento", {})
        linee.append(f"{key}: confronto M1/riferimento, testo uguale {cmp.get('testo_uguale')}/"
                     f"{cmp.get('prompt_comuni')}, categoria uguale {cmp.get('categoria_uguale')}/"
                     f"{cmp.get('prompt_comuni')}")
        acc = c["accuratezza_B_sintetico"].get("tutte", {}).get("accuratezza")
        linee.append(f"{key}: accuratezza su B {acc if acc is None else round(acc, 3)}"
                     + (" (SOTTO 0.80: il modello legge male i passaggi)" if c["HC0_set_avviso"] else "")
                     + f"; «altro» {c['quota_altro_sintetico']:.1%}"
                     + (" — OLTRE IL 10%: rivedere le regole sui soli dati della prova" if SMOKE and
                        (c["quota_altro_sintetico"] or 0) > SOGLIA_ALTRO_PROVA else "")
                     + f"; risposte finite entro {TOKEN_RISPOSTA} token {c['risposte_finite_entro_il_limite']:.1%}")
    linee.append("")
    for r in res["risultati"]:
        h4 = r.get("HC4", {})
        h0 = r["HC0"] or {}
        h0_txt = h0.get("esito") or ("mancante" if not h0 else ("superato" if h0.get("superato") else "non superato"))
        if h0.get("valore") is not None:
            h0_txt += f" ({h0['valore']:.3f})"
        linee.append(f"{r['metodo']} · {r['modello']}: HC0 {h0_txt}"
                     f" | W {stat_txt(r['W'])} → HC1 {r['HC1']} | ΔW {stat_txt(r.get('delta_W'))} → HC2 {r.get('HC2')}"
                     f" | W_RIM {stat_txt(r['W_RIM'])} | D {stat_txt(r['D_RIM'])} → HC3 {r['HC3']}"
                     f" | HC4 C {stat_txt(h4.get('entro_C'))} ({h4.get('entro_C', {}).get('n_pos')}/"
                     f"{h4.get('entro_C', {}).get('n_neg')}) → {h4.get('entro_C', {}).get('esito')}"
                     f" | HC4 B {stat_txt(h4.get('entro_B'))} ({h4.get('entro_B', {}).get('n_pos')}/"
                     f"{h4.get('entro_B', {}).get('n_neg')}) → {h4.get('entro_B', {}).get('esito')}")
    linee += ["", "Lettura:"] + (["Prova tecnica: nessun esito delle ipotesi; si controllano codice, classificazione e tempi "
                                  "(§8)."] if SMOKE else []) + out["lettura"]
    if res["errori"]:
        linee += ["", "ERRORI NELL'ANALISI:"] + res["errori"]
    (RADICE / "rapporto_C.txt").write_text("\n".join(linee))
    print("\n".join(linee))
    zpath = RADICE / f"testC_{MODALITA}_risultati.zip"
    with zipfile.ZipFile(zpath, "w", zipfile.ZIP_DEFLATED) as z:
        for name in ("esiti_C.json", "tabella_C.csv", "rapporto_C.txt", "risposte_C.csv", "campione_classificazione.csv",
                     "campione_classificazione_chiave.json", "confronto_M1_riferimento.csv", "stato.json"):
            if (RADICE / name).is_file():
                z.write(RADICE / name, name)
        for f in sorted((RADICE / "log").glob("*.log")):
            z.write(f, f"log/{f.name}")
        for f in sorted((RADICE / "ambienti").glob("*.txt")):
            z.write(f, f"ambienti/{f.name}")
        for job, info in st.items():
            fp = job_file(job)
            if fp is not None:
                z.write(fp, f"uscite/{fp.name}")
        config_export = {k: globals()[k] for k in ("VERSIONE", "MODALITA", "MODELLI", "METODI", "M1_GENERAZIONE", "TOKEN_RISPOSTA", "SEED", "SEME_METODI", "N_BOOT", "SHA_RIGHE_C", "SHA_SINTETICO", "SIRIN_COMMIT", "OBESO_COMMIT", "SONDE_SHA", "GIUDICE_SHA", "MIN_RISPOSTE_HC4", "MIN_PASSAGGI_HC4", "QUOTA_COPERTURA", "REVISIONE_DATI_APPROVATA", "VERSIONE_SELEZIONE", "REGOLE_NATURALE", "SHA_ARCHIVIO_SELEZIONE")}
        S.bundle_inputs(z, RIGHE, PASSAGGI, DATI, st, res["hc0_fonti"], config_export, SCRIPT)
    print("\nZIP da mandare:", zpath)
    return zpath
```
